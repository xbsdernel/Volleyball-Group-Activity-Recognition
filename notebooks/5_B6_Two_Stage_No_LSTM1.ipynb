# Model 5: Baseline 6 (B6) â€” Two-stage Model without LSTM 1
**Reference:** Ibrahim et al., *A Hierarchical Deep Temporal Model for Group Activity Recognition*, CVPR 2016 (Section 4.1 #6)

### Architecture & Workflow
"We train [person action fine-tuned CNN] features for individual people and pool them together to represent each frame and feed them to a group LSTM."

This is a **variant of the full hierarchical model that removes LSTM 1** (the person-level temporal model). The fine-tuned person CNN from B3 provides features; those are max-pooled across players per frame and fed to a group-level LSTM 2.

**Data Flow:**
$$\underbrace{9 \text{ frames} \times 12 \text{ crops}}_{\text{input}} \xrightarrow{\text{Fine-tuned CNN}} (B, 9, 12, 2048) \xrightarrow{\max_k} (B, 9, 2048) \xrightarrow{\text{LSTM 2}} (B, 500) \xrightarrow{\text{Linear}} 8 \text{ classes}$$

**Key difference from B5:** B5 uses off-the-shelf ImageNet CNN features. B6 uses a **fine-tuned** person action CNN (Stage 1 from B3).
**Key difference from full Hierarchical Model:** No LSTM 1 between the person CNN and the group LSTM 2.

```python
import os
import json
import random
import warnings
from PIL import Image
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm.auto import tqdm

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader, TensorDataset
import torchvision.transforms as transforms
from torchvision import models
from sklearn.metrics import accuracy_score, f1_score, precision_score, recall_score, confusion_matrix, classification_report

warnings.filterwarnings('ignore')

DATASET_ROOT  = "/kaggle/input/datasets/ahmedmohamed365/volleyball"
VIDEOS_ROOT   = "/kaggle/input/datasets/ahmedmohamed365/volleyball/volleyball_/videos"
TRACKING_ROOT = "/kaggle/input/datasets/ahmedmohamed365/volleyball/volleyball_tracking_annotation/volleyball_tracking_annotation"

# Results directory setup
RESULTS_DIR = "results/5_B6_Two_Stage_No_LSTM1"
PLOTS_DIR = os.path.join(RESULTS_DIR, "plots")
METRICS_DIR = os.path.join(RESULTS_DIR, "metrics")
os.makedirs(PLOTS_DIR, exist_ok=True)
os.makedirs(METRICS_DIR, exist_ok=True)

SEED = 42
random.seed(SEED); np.random.seed(SEED)
torch.manual_seed(SEED); torch.cuda.manual_seed_all(SEED)
torch.backends.cudnn.benchmark = True

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Device: {device}")
```

```python
TRAIN_VIDEOS = {1, 3, 6, 7, 10, 13, 15, 16, 18, 22, 23, 31, 32, 36, 38, 39, 40, 41, 42, 48, 50, 52, 53, 54}
VAL_VIDEOS   = {0, 2, 8, 12, 17, 19, 24, 26, 27, 28, 30, 33, 46, 49, 51}
TEST_VIDEOS  = {4, 5, 9, 11, 14, 20, 21, 25, 29, 34, 35, 37, 43, 44, 45, 47}

LABEL_MAP = {
    "l_pass": 0, "r_pass": 1, "l_spike": 2, "r_spike": 3,
    "l_set": 4, "r_set": 5, "l_winpoint": 6, "r_winpoint": 7,
    "l-pass": 0, "r-pass": 1, "l-spike": 2, "r-spike": 3,
    "l-set": 4, "r-set": 5, "l-winpoint": 6, "r-winpoint": 7
}

PERSON_ACTION_MAP = {
    "waiting": 0, "setting": 1, "digging": 2, "falling": 3,
    "spiking": 4, "blocking": 5, "jumping": 6, "moving": 7, "standing": 8
}

IMAGE_SIZE = (224, 224)
BATCH_SIZE = 16
LR = 1e-3
EPOCHS_STAGE1 = 15
EPOCHS_STAGE2 = 40
NUM_FRAMES = 9
NUM_PLAYERS = 12
```

```python
# Tracking Annotation Parser
class BoxInfo:
    def __init__(self, line):
        words = line.split()
        self.category = words.pop()
        words = [int(w) for w in words]
        self.player_ID = words[0]
        x1, y1, x2, y2, frame_ID, lost, grouping, generated = words[1:]
        self.box = (x1, y1, x2, y2)
        self.frame_ID = frame_ID

def load_tracking_annot(path):
    player_boxes = {idx: [] for idx in range(12)}
    frame_boxes_dct = {}
    if not os.path.isfile(path):
        return frame_boxes_dct
    with open(path, 'r') as f:
        for line in f:
            b = BoxInfo(line)
            if b.player_ID > 11:
                continue
            player_boxes[b.player_ID].append(b)
    for pid, boxes in player_boxes.items():
        boxes = boxes[5:-6] if len(boxes) > 11 else boxes
        for b in boxes:
            frame_boxes_dct.setdefault(b.frame_ID, []).append(b)
    return frame_boxes_dct

def get_sequence_frame_paths(mid_img_path, n=4):
    clip_dir = os.path.dirname(mid_img_path)
    name, ext = os.path.splitext(os.path.basename(mid_img_path))
    mid_id = int(name)
    return [p if os.path.isfile(p := os.path.join(clip_dir, f"{mid_id+o}{ext}")) else mid_img_path
            for o in range(-n, n+1)]
```

```python
# Load Full Dataset (All 55 Videos)
data = []
for vid_name in sorted(os.listdir(VIDEOS_ROOT)):
    vpath = os.path.join(VIDEOS_ROOT, vid_name)
    if not os.path.isdir(vpath): continue
    ann_path = os.path.join(vpath, "annotations.txt")
    if not os.path.isfile(ann_path): continue
    vid_id = int(vid_name)
    with open(ann_path, 'r') as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) >= 2:
                img_name, grp_lbl = parts[0], parts[1]
                clip_id = os.path.splitext(img_name)[0]
                mid_img_path = os.path.join(vpath, clip_id, img_name)
                trk_path = os.path.join(TRACKING_ROOT, str(vid_id), clip_id, f"{clip_id}.txt")
                seq_paths = get_sequence_frame_paths(mid_img_path, n=4)
                data.append({"video_id": vid_id, "frame_paths": seq_paths, "tracking_path": trk_path, "group_label": grp_lbl, "image_path": mid_img_path})

train_data = [d for d in data if d["video_id"] in TRAIN_VIDEOS]
val_data   = [d for d in data if d["video_id"] in VAL_VIDEOS]
test_data  = [d for d in data if d["video_id"] in TEST_VIDEOS]
print(f"Total clips: {len(data)} | Train: {len(train_data)} | Val: {len(val_data)} | Test: {len(test_data)}")
```

```python
# Transforms
crop_transform = transforms.Compose([
    transforms.Resize(IMAGE_SIZE),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

# Stage 1: Single-crop Dataset for Person Action Fine-tuning
class SinglePlayerCropDataset(Dataset):
    def __init__(self, data, transform=None):
        self.samples = []
        self.transform = transform
        for d in data:
            if not os.path.isfile(d["image_path"]) or not os.path.isfile(d["tracking_path"]):
                continue
            fboxes = load_tracking_annot(d["tracking_path"])
            if not fboxes: continue
            fid = list(fboxes.keys())[0]
            for b in fboxes[fid][:12]:
                act_idx = PERSON_ACTION_MAP.get(b.category.lower(), 8)
                self.samples.append((d["image_path"], b.box, act_idx))

    def __len__(self): return len(self.samples)

    def __getitem__(self, idx):
        img_path, (x1, y1, x2, y2), act_idx = self.samples[idx]
        img = Image.open(img_path).convert("RGB")
        w, h = img.size
        crop = img.crop((max(0,x1), max(0,y1), min(w, max(x1+1,x2)), min(h, max(y1+1,y2))))
        if self.transform: crop = self.transform(crop)
        return crop, torch.tensor(act_idx, dtype=torch.long)

# Stage 2: Multi-frame Multi-player Dataset
class VolleyballSequencePersonDataset(Dataset):
    """Returns (9, 12, 3, 224, 224) clip tensor and group label."""
    def __init__(self, data, transform=None):
        self.data = data
        self.transform = transform

    def __len__(self): return len(self.data)

    def __getitem__(self, idx):
        d = self.data[idx]
        fboxes = load_tracking_annot(d["tracking_path"])
        all_frames = []
        for p in d["frame_paths"]:
            img = Image.open(p).convert("RGB")
            fid_str = os.path.splitext(os.path.basename(p))[0]
            try: fid = int(fid_str)
            except: fid = list(fboxes.keys())[0] if fboxes else 0
            boxes = fboxes.get(fid, [])
            crops = []
            w, h = img.size
            for b in boxes[:NUM_PLAYERS]:
                x1, y1, x2, y2 = b.box
                c = img.crop((max(0,x1), max(0,y1), min(w,max(x1+1,x2)), min(h,max(y1+1,y2))))
                if self.transform: c = self.transform(c)
                crops.append(c)
            while len(crops) < NUM_PLAYERS:
                crops.append(torch.zeros(3, *IMAGE_SIZE))
            all_frames.append(torch.stack(crops))  # (12, 3, H, W)
        clip = torch.stack(all_frames)  # (9, 12, 3, H, W)
        label = torch.tensor(LABEL_MAP[d["group_label"]], dtype=torch.long)
        return clip, label
```

```python
# Stage 1: Fine-tune Person Action CNN
class PersonActionCNN(nn.Module):
    def __init__(self, num_actions=9):
        super().__init__()
        resnet = models.resnet50(weights=models.ResNet50_Weights.DEFAULT)
        self.backbone = nn.Sequential(*list(resnet.children())[:-1])
        self.classifier = nn.Linear(2048, num_actions)

    def forward(self, x):
        f = self.backbone(x).view(x.size(0), -1)
        return self.classifier(f)

stage1_model = PersonActionCNN(num_actions=9).to(device)
stage1_train_ds = SinglePlayerCropDataset(train_data, transform=crop_transform)
stage1_loader   = DataLoader(stage1_train_ds, batch_size=32, shuffle=True, num_workers=2, pin_memory=True)

optimizer1 = optim.AdamW(stage1_model.parameters(), lr=1e-4, weight_decay=1e-4)
criterion1 = nn.CrossEntropyLoss()
scaler = torch.cuda.amp.GradScaler(enabled=(device == "cuda"))

print(f"Stage 1 person-action crops available: {len(stage1_train_ds)}")
print("Stage 1 training loop â€” uncomment to train Stage 1 from scratch:")
# for epoch in range(EPOCHS_STAGE1):
#     stage1_model.train()
#     for crops, acts in tqdm(stage1_loader, leave=False):
#         crops, acts = crops.to(device), acts.to(device)
#         optimizer1.zero_grad()
#         with torch.cuda.amp.autocast(enabled=(device=="cuda")):
#             loss = criterion1(stage1_model(crops), acts)
#         scaler.scale(loss).backward()
#         scaler.step(optimizer1); scaler.update()
# torch.save(stage1_model.state_dict(), "/kaggle/working/person_action_cnn_b6.pth")
```

```python
# Stage 2: B6 Model â€” Fine-tuned CNN Features + Group LSTM 2 (No LSTM 1)
class B6Model(nn.Module):
    """
    B6: Two-stage model WITHOUT LSTM 1.
    Fine-tuned person CNN backbone (frozen) -> max-pool over players per frame -> Group LSTM 2.
    """
    def __init__(self, finetuned_backbone, num_classes=8, hidden_dim=500, dropout=0.5):
        super().__init__()
        self.backbone = finetuned_backbone
        for p in self.backbone.parameters():
            p.requires_grad = False  # Frozen fine-tuned person CNN

        # Group LSTM 2 only (no person-level LSTM 1)
        self.lstm2 = nn.LSTM(input_size=2048, hidden_size=hidden_dim, num_layers=1, batch_first=True)
        self.dropout = nn.Dropout(dropout)
        self.classifier = nn.Linear(hidden_dim, num_classes)

    def forward(self, x):
        # x: (B, 9, 12, 3, 224, 224)
        B, T, K, C, H, W = x.shape
        x_flat = x.view(B * T * K, C, H, W)
        with torch.no_grad():
            feats = self.backbone(x_flat).view(B, T, K, 2048)
        # Max-pool over players per frame
        scene_seq = feats.max(dim=2)[0]  # (B, 9, 2048)
        # Group LSTM 2
        _, (hn, _) = self.lstm2(scene_seq)
        return self.classifier(self.dropout(hn[-1]))

b6_model = B6Model(finetuned_backbone=stage1_model.backbone, num_classes=8).to(device)

# Only train LSTM 2 + classifier (backbone frozen)
optimizer2 = optim.AdamW(
    list(b6_model.lstm2.parameters()) + list(b6_model.classifier.parameters()),
    lr=LR, weight_decay=1e-4
)
criterion2 = nn.CrossEntropyLoss()
scaler2 = torch.cuda.amp.GradScaler(enabled=(device == "cuda"))
scheduler2 = optim.lr_scheduler.CosineAnnealingLR(optimizer2, T_max=EPOCHS_STAGE2, eta_min=1e-6)

print("B6 Model ready.")
trainable_params = sum(p.numel() for p in b6_model.parameters() if p.requires_grad)
print(f"Trainable parameters (LSTM2 + head only): {trainable_params:,}")
```

```python
# DataLoaders for Stage 2
b6_train_ds = VolleyballSequencePersonDataset(train_data, transform=crop_transform)
b6_val_ds   = VolleyballSequencePersonDataset(val_data,   transform=crop_transform)
b6_test_ds  = VolleyballSequencePersonDataset(test_data,  transform=crop_transform)

loader_kw = dict(num_workers=2, pin_memory=(device == "cuda"), persistent_workers=True)
b6_train_loader = DataLoader(b6_train_ds, batch_size=BATCH_SIZE, shuffle=True, drop_last=True, **loader_kw)
b6_val_loader   = DataLoader(b6_val_ds,   batch_size=BATCH_SIZE, shuffle=False, **loader_kw)
b6_test_loader  = DataLoader(b6_test_ds,  batch_size=BATCH_SIZE, shuffle=False, **loader_kw)
print(f"Train batches: {len(b6_train_loader)} | Val batches: {len(b6_val_loader)} | Test batches: {len(b6_test_loader)}")
```

```python
# Stage 2 Training Loop
history = {"train_loss": [], "val_loss": [], "train_acc": [], "val_acc": []}
best_val_acc = 0.0

for epoch in range(EPOCHS_STAGE2):
    b6_model.train()
    total_loss, correct, total = 0.0, 0, 0
    for inputs, labels in tqdm(b6_train_loader, desc=f"Ep {epoch+1:02d}/{EPOCHS_STAGE2} [Train]", leave=False):
        inputs, labels = inputs.to(device, non_blocking=True), labels.to(device, non_blocking=True)
        optimizer2.zero_grad()
        with torch.cuda.amp.autocast(enabled=(device == "cuda")):
            out = b6_model(inputs)
            loss = criterion2(out, labels)
        scaler2.scale(loss).backward()
        scaler2.step(optimizer2); scaler2.update()
        total_loss += loss.item() * inputs.size(0)
        correct += (out.argmax(1) == labels).sum().item()
        total += labels.size(0)

    tr_loss, tr_acc = total_loss / total, correct / total * 100

    b6_model.eval()
    total_loss, correct, total = 0.0, 0, 0
    with torch.no_grad():
        for inputs, labels in tqdm(b6_val_loader, desc=f"Ep {epoch+1:02d} [Val]", leave=False):
            inputs, labels = inputs.to(device, non_blocking=True), labels.to(device, non_blocking=True)
            with torch.cuda.amp.autocast(enabled=(device == "cuda")):
                out = b6_model(inputs)
                loss = criterion2(out, labels)
            total_loss += loss.item() * inputs.size(0)
            correct += (out.argmax(1) == labels).sum().item()
            total += labels.size(0)

    val_loss, val_acc = total_loss / total, correct / total * 100
    scheduler2.step()

    history["train_loss"].append(tr_loss); history["val_loss"].append(val_loss)
    history["train_acc"].append(tr_acc);   history["val_acc"].append(val_acc)
    print(f"Epoch {epoch+1:02d}/{EPOCHS_STAGE2} | Train Loss: {tr_loss:.4f} Acc: {tr_acc:.2f}% | Val Loss: {val_loss:.4f} Acc: {val_acc:.2f}%")

    if val_acc > best_val_acc:
        best_val_acc = val_acc
        torch.save(b6_model.state_dict(), "/kaggle/working/b6_best.pth")
torch.save(b6_model.state_dict(), os.path.join(RESULTS_DIR, "best_model.pth"))
        print(f"  --> Saved best B6 checkpoint (Val Acc: {best_val_acc:.2f}%)")
```

```python
# Final Test Evaluation & Metrics Export
best_ckpt = os.path.join(RESULTS_DIR, "best_model.pth")
if not os.path.isfile(best_ckpt):
    best_ckpt = "/kaggle/working/b6_best.pth"
if os.path.isfile(best_ckpt):
    b6_model.load_state_dict(torch.load(best_ckpt, map_location=device))

b6_model.eval()
preds, targets = [], []
with torch.no_grad():
    for inputs, labels in b6_test_loader:
        inputs = inputs.to(device, non_blocking=True)
        with torch.cuda.amp.autocast(enabled=(device == "cuda")):
            out = b6_model(inputs)
        preds.extend(out.argmax(1).cpu().numpy())
        targets.extend(labels.numpy())

class_names = ["l_pass", "r_pass", "l_spike", "r_spike", "l_set", "r_set", "l_winpoint", "r_winpoint"]
test_acc = accuracy_score(targets, preds)
test_f1_weighted = f1_score(targets, preds, average='weighted', zero_division=0)
test_f1_macro = f1_score(targets, preds, average='macro', zero_division=0)
test_prec = precision_score(targets, preds, average='weighted', zero_division=0)
test_rec = recall_score(targets, preds, average='weighted', zero_division=0)

metrics_dict = {
    "model": "B6 - Two-stage Model without LSTM 1",
    "test_accuracy": float(test_acc),
    "test_f1_weighted": float(test_f1_weighted),
    "test_f1_macro": float(test_f1_macro),
    "test_precision": float(test_prec),
    "test_recall": float(test_rec),
    "best_val_accuracy": float(best_val_acc)
}

with open(os.path.join(METRICS_DIR, "metrics.json"), 'w') as f:
    json.dump(metrics_dict, f, indent=4)

pd.DataFrame(history).to_csv(os.path.join(METRICS_DIR, "training_history.csv"), index=False)

rep_dict = classification_report(targets, preds, target_names=class_names, output_dict=True, zero_division=0)
pd.DataFrame(rep_dict).transpose().to_csv(os.path.join(METRICS_DIR, "classification_report.csv"))

print(f"Test Accuracy: {test_acc*100:.2f}%")
print(f"Test F1 (weighted): {test_f1_weighted:.4f}")
print(classification_report(targets, preds, target_names=class_names, zero_division=0))
```

```python
# Training Curves & Confusion Matrix (Display and Save)
fig, axs = plt.subplots(1, 2, figsize=(14, 5))
axs[0].plot(history["train_loss"], label="Train Loss", color="royalblue")
axs[0].plot(history["val_loss"], label="Val Loss", color="crimson")
axs[0].set_title("B6 Loss"); axs[0].legend(); axs[0].grid(alpha=0.3)
axs[1].plot(history["train_acc"], label="Train Acc", color="royalblue")
axs[1].plot(history["val_acc"], label="Val Acc", color="crimson")
axs[1].set_title("B6 Accuracy"); axs[1].legend(); axs[1].grid(alpha=0.3)
plt.tight_layout()
plt.savefig(os.path.join(PLOTS_DIR, "training_curves.png"), dpi=300)
plt.show()

cm = confusion_matrix(targets, preds)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=class_names, yticklabels=class_names)
plt.title('B6 Confusion Matrix (Test Split)')
plt.xlabel('Predicted Label')
plt.ylabel('Ground Truth Label')
plt.tight_layout()
plt.savefig(os.path.join(PLOTS_DIR, "confusion_matrix.png"), dpi=300)
plt.show()
```
