# Model 2: Baseline 3 (B3) â€” Fine-tuned Person Classification
**Reference:** Ibrahim et al., *A Hierarchical Deep Temporal Model for Group Activity Recognition*, CVPR 2016 (Section 4.1 #3)

### Architecture & Workflow
1. **Stage 1 (Person Action Fine-Tuning):** A ResNet-50 backbone is fine-tuned on individual player bounding-box crops to recognize person-level actions (9 classes).
2. **Stage 2 (Group Activity Recognition):** The fine-tuned ResNet-50 backbone is frozen (without further fine-tuning). For each frame, 2048-dim features are extracted for all 12 players, **max-pooled** across players as specified in the paper (Section 3.2, 4.1, 4.3):
   $$Z = \max_{k=1..12} x_k \in \mathbb{R}^{2048}$$
3. A linear/softmax classifier ($2048 \rightarrow 8$) is trained on the pooled feature $Z$ to predict the group activity.

```python
import os
import json
import random
import warnings
from PIL import Image
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm.auto import tqdm

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader, TensorDataset
import torchvision.transforms as transforms
from torchvision import models
from sklearn.metrics import accuracy_score, f1_score, precision_score, recall_score, confusion_matrix, classification_report

warnings.filterwarnings('ignore')

# Paths & Hardware Setup
DATASET_ROOT = "/kaggle/input/datasets/ahmedmohamed365/volleyball"
VIDEOS_ROOT = "/kaggle/input/datasets/ahmedmohamed365/volleyball/volleyball_/videos"
TRACKING_ROOT = "/kaggle/input/datasets/ahmedmohamed365/volleyball/volleyball_tracking_annotation/volleyball_tracking_annotation"

# Results directory setup
RESULTS_DIR = "results/2_B3_Finetuned_Person_Classification"
PLOTS_DIR = os.path.join(RESULTS_DIR, "plots")
METRICS_DIR = os.path.join(RESULTS_DIR, "metrics")
os.makedirs(PLOTS_DIR, exist_ok=True)
os.makedirs(METRICS_DIR, exist_ok=True)

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)
torch.backends.cudnn.benchmark = True

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Device: {device}")
```

```python
# Dataset Configuration & Splits
TRAIN_VIDEOS = {1, 3, 6, 7, 10, 13, 15, 16, 18, 22, 23, 31, 32, 36, 38, 39, 40, 41, 42, 48, 50, 52, 53, 54}
VAL_VIDEOS   = {0, 2, 8, 12, 17, 19, 24, 26, 27, 28, 30, 33, 46, 49, 51}
TEST_VIDEOS  = {4, 5, 9, 11, 14, 20, 21, 25, 29, 34, 35, 37, 43, 44, 45, 47}

LABEL_MAP = {
    "l_pass": 0, "r_pass": 1, "l_spike": 2, "r_spike": 3,
    "l_set": 4, "r_set": 5, "l_winpoint": 6, "r_winpoint": 7,
    "l-pass": 0, "r-pass": 1, "l-spike": 2, "r-spike": 3,
    "l-set": 4, "r-set": 5, "l-winpoint": 6, "r-winpoint": 7
}

PERSON_ACTION_MAP = {
    "waiting": 0, "setting": 1, "digging": 2, "falling": 3,
    "spiking": 4, "blocking": 5, "jumping": 6, "moving": 7, "standing": 8
}

IMAGE_SIZE = (224, 224)
BATCH_SIZE = 16
LR = 1e-3
EPOCHS_STAGE1 = 15
EPOCHS_STAGE2 = 30
```

```python
# Tracking Annotation Parser
class BoxInfo:
    def __init__(self, line):
        words = line.split()
        self.category = words.pop()
        words = [int(w) for w in words]
        self.player_ID = words[0]
        x1, y1, x2, y2, frame_ID, lost, grouping, generated = words[1:]
        self.box = (x1, y1, x2, y2)
        self.frame_ID = frame_ID

def load_tracking_annot(path):
    player_boxes = {idx: [] for idx in range(12)}
    frame_boxes_dct = {}
    if not os.path.isfile(path):
        return frame_boxes_dct
    with open(path, 'r') as file:
        for line in file:
            box_info = BoxInfo(line)
            if box_info.player_ID > 11:
                continue
            player_boxes[box_info.player_ID].append(box_info)

    for player_ID, boxes_info in player_boxes.items():
        boxes_info = boxes_info[5:-6] if len(boxes_info) > 11 else boxes_info
        for box_info in boxes_info:
            if box_info.frame_ID not in frame_boxes_dct:
                frame_boxes_dct[box_info.frame_ID] = []
            frame_boxes_dct[box_info.frame_ID].append(box_info)
    return frame_boxes_dct
```

```python
# Load Full Dataset Annotations (All 55 videos)
data = []
for vid_name in sorted(os.listdir(VIDEOS_ROOT)):
    vpath = os.path.join(VIDEOS_ROOT, vid_name)
    if not os.path.isdir(vpath):
        continue
    ann_path = os.path.join(vpath, "annotations.txt")
    if not os.path.isfile(ann_path):
        continue
    vid_id = int(vid_name)
    with open(ann_path, 'r') as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) >= 2:
                img_name = parts[0]
                grp_lbl = parts[1]
                clip_id = os.path.splitext(img_name)[0]
                img_path = os.path.join(vpath, clip_id, img_name)
                trk_path = os.path.join(TRACKING_ROOT, str(vid_id), clip_id, f"{clip_id}.txt")
                data.append({"video_id": vid_id, "clip_id": clip_id, "image_path": img_path, "tracking_path": trk_path, "group_label": grp_lbl})

train_data = [d for d in data if d["video_id"] in TRAIN_VIDEOS]
val_data   = [d for d in data if d["video_id"] in VAL_VIDEOS]
test_data  = [d for d in data if d["video_id"] in TEST_VIDEOS]
print(f"Total clips: {len(data)} | Train: {len(train_data)} | Val: {len(val_data)} | Test: {len(test_data)}")
```

```python
# Stage 1 & Stage 2 Dataset Definitions
crop_transform = transforms.Compose([
    transforms.Resize(IMAGE_SIZE),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

class SinglePlayerCropDataset(Dataset):
    """Extracts individual player crops annotated with person actions for Stage 1 fine-tuning."""
    def __init__(self, data, transform=None):
        self.samples = []
        self.transform = transform
        for d in data:
            if not os.path.isfile(d["image_path"]) or not os.path.isfile(d["tracking_path"]):
                continue
            fboxes = load_tracking_annot(d["tracking_path"])
            if not fboxes:
                continue
            fid = list(fboxes.keys())[0]
            for b in fboxes[fid][:12]:
                act_idx = PERSON_ACTION_MAP.get(b.category.lower(), 8)
                self.samples.append((d["image_path"], b.box, act_idx))

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        img_path, (x1, y1, x2, y2), act_idx = self.samples[idx]
        img = Image.open(img_path).convert("RGB")
        w, h = img.size
        crop = img.crop((max(0, x1), max(0, y1), min(w, max(x1+1, x2)), min(h, max(y1+1, y2))))
        if self.transform:
            crop = self.transform(crop)
        return crop, torch.tensor(act_idx, dtype=torch.long)

class Volleyball12PlayersDataset(Dataset):
    """Extracts all 12 player crops for group activity recognition (Stage 2)."""
    def __init__(self, data, transform=None):
        self.data = data
        self.transform = transform

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        d = self.data[idx]
        img = Image.open(d["image_path"]).convert("RGB")
        fboxes = load_tracking_annot(d["tracking_path"])
        fid = list(fboxes.keys())[0] if fboxes else 0
        boxes = fboxes.get(fid, [])
        crops = []
        w, h = img.size
        for b in boxes[:12]:
            x1, y1, x2, y2 = b.box
            c = img.crop((max(0, x1), max(0, y1), min(w, max(x1+1, x2)), min(h, max(y1+1, y2))))
            if self.transform:
                c = self.transform(c)
            crops.append(c)
        while len(crops) < 12:
            crops.append(torch.zeros(3, *IMAGE_SIZE))
        return torch.stack(crops), torch.tensor(LABEL_MAP[d["group_label"]], dtype=torch.long)
```

```python
# Model Definition: Stage 1 (Person Action CNN) & Stage 2 (B3Model with MAX POOLING)
class PersonActionCNN(nn.Module):
    def __init__(self, num_actions=9):
        super().__init__()
        resnet = models.resnet50(weights=models.ResNet50_Weights.DEFAULT)
        self.backbone = nn.Sequential(*list(resnet.children())[:-1])
        self.classifier = nn.Linear(2048, num_actions)

    def forward(self, x):
        feats = self.backbone(x).view(x.size(0), -1)
        return self.classifier(feats)

class B3Model(nn.Module):
    """Baseline 3: Frozen fine-tuned person CNN -> Max-pooling over players -> Group classifier."""
    def __init__(self, fine_tuned_backbone, num_classes=8):
        super().__init__()
        self.backbone = fine_tuned_backbone
        for p in self.backbone.parameters():
            p.requires_grad = False  # Backbone frozen in Stage 2 as in paper
        self.classifier = nn.Linear(2048, num_classes)

    def forward(self, x):
        # x: (B, 12, 3, 224, 224)
        B, K, C, H, W = x.shape
        x_flat = x.view(B * K, C, H, W)
        with torch.no_grad():
            feats = self.backbone(x_flat).view(B, K, 2048)
        # MAX-POOLING over all 12 players: Z = max_k (x_k) [Paper Section 3.2, 4.1, 4.3]
        z = feats.max(dim=1)[0]  # (B, 2048)
        return self.classifier(z)
```

### Stage 1: Fine-tuning ResNet-50 on Individual Person Actions
Trains the CNN on bounding box crops using individual player action annotations (9 classes).

```python
stage1_train_ds = SinglePlayerCropDataset(train_data, transform=crop_transform)
stage1_val_ds   = SinglePlayerCropDataset(val_data, transform=crop_transform)
print(f"Stage 1 Person Action Crops -> Train: {len(stage1_train_ds)} | Val: {len(stage1_val_ds)}")

stage1_loader = DataLoader(stage1_train_ds, batch_size=32, shuffle=True, num_workers=2, pin_memory=True)
stage1_model = PersonActionCNN(num_actions=9).to(device)

optimizer1 = optim.AdamW(stage1_model.parameters(), lr=1e-4, weight_decay=1e-4)
criterion1 = nn.CrossEntropyLoss()
scaler = torch.cuda.amp.GradScaler(enabled=(device == "cuda"))

print("Training Stage 1 (Person Action CNN)... (Can load pretrained checkpoint if available)")
# for epoch in range(EPOCHS_STAGE1):
#     stage1_model.train()
#     for crops, acts in tqdm(stage1_loader, leave=False):
#         crops, acts = crops.to(device), acts.to(device)
#         optimizer1.zero_grad()
#         with torch.cuda.amp.autocast(enabled=(device == "cuda")):
#             out = stage1_model(crops)
#             loss = criterion1(out, acts)
#         scaler.scale(loss).backward()
#         scaler.step(optimizer1)
#         scaler.update()
```

### Stage 2: Group Activity Recognition (Max-pooling over players + Group Classifier)
Freeze the fine-tuned person CNN, extract features, apply max-pooling, and train the group activity classifier.

```python
b3_train_ds = Volleyball12PlayersDataset(train_data, transform=crop_transform)
b3_val_ds   = Volleyball12PlayersDataset(val_data, transform=crop_transform)
b3_test_ds  = Volleyball12PlayersDataset(test_data, transform=crop_transform)

loader_kw = dict(num_workers=2, pin_memory=(device == "cuda"), persistent_workers=True)
b3_train_loader = DataLoader(b3_train_ds, batch_size=BATCH_SIZE, shuffle=True, drop_last=True, **loader_kw)
b3_val_loader   = DataLoader(b3_val_ds, batch_size=BATCH_SIZE, shuffle=False, **loader_kw)
b3_test_loader  = DataLoader(b3_test_ds, batch_size=BATCH_SIZE, shuffle=False, **loader_kw)

b3_model = B3Model(fine_tuned_backbone=stage1_model.backbone, num_classes=8).to(device)
optimizer2 = optim.AdamW(b3_model.classifier.parameters(), lr=LR, weight_decay=1e-4)
criterion2 = nn.CrossEntropyLoss()
scaler2 = torch.cuda.amp.GradScaler(enabled=(device == "cuda"))
scheduler2 = optim.lr_scheduler.CosineAnnealingLR(optimizer2, T_max=EPOCHS_STAGE2, eta_min=1e-6)

history = {'train_loss': [], 'val_loss': [], 'train_acc': [], 'val_acc': []}
best_val_acc = 0.0

print(f"Starting Stage 2 training for {EPOCHS_STAGE2} epochs...")
for epoch in range(EPOCHS_STAGE2):
    b3_model.train()
    tr_loss, tr_corr, tr_tot = 0.0, 0, 0
    for x, y in tqdm(b3_train_loader, desc=f"Ep {epoch+1:02d}/{EPOCHS_STAGE2} [Train]", leave=False):
        x, y = x.to(device, non_blocking=True), y.to(device, non_blocking=True)
        optimizer2.zero_grad()
        with torch.cuda.amp.autocast(enabled=(device == "cuda")):
            out = b3_model(x)
            loss = criterion2(out, y)
        scaler2.scale(loss).backward()
        scaler2.step(optimizer2); scaler2.update()
        tr_loss += loss.item() * x.size(0)
        tr_corr += (out.argmax(1) == y).sum().item()
        tr_tot += y.size(0)

    b3_model.eval()
    val_loss, val_corr, val_tot = 0.0, 0, 0
    with torch.no_grad():
        for x, y in tqdm(b3_val_loader, desc=f"Ep {epoch+1:02d} [Val]", leave=False):
            x, y = x.to(device, non_blocking=True), y.to(device, non_blocking=True)
            with torch.cuda.amp.autocast(enabled=(device == "cuda")):
                out = b3_model(x)
                loss = criterion2(out, y)
            val_loss += loss.item() * x.size(0)
            val_corr += (out.argmax(1) == y).sum().item()
            val_tot += y.size(0)

    scheduler2.step()
    ep_tr_loss, ep_tr_acc = tr_loss / tr_tot, tr_corr / tr_tot * 100
    ep_val_loss, ep_val_acc = val_loss / val_tot, val_corr / val_tot * 100
    history['train_loss'].append(ep_tr_loss)
    history['val_loss'].append(ep_val_loss)
    history['train_acc'].append(ep_tr_acc)
    history['val_acc'].append(ep_val_acc)
    print(f"Epoch {epoch+1:02d}/{EPOCHS_STAGE2} | Train Loss: {ep_tr_loss:.4f} Acc: {ep_tr_acc:.2f}% | Val Loss: {ep_val_loss:.4f} Acc: {ep_val_acc:.2f}%")

    if ep_val_acc > best_val_acc:
        best_val_acc = ep_val_acc
        torch.save(b3_model.state_dict(), os.path.join(RESULTS_DIR, "best_model.pth"))
        print(f"  --> Saved best B3 checkpoint (Val Acc: {best_val_acc:.2f}%)")

# Evaluation on Test Split
best_path = os.path.join(RESULTS_DIR, "best_model.pth")
if os.path.isfile(best_path):
    b3_model.load_state_dict(torch.load(best_path, map_location=device))

b3_model.eval()
test_preds, test_targets = [], []
with torch.no_grad():
    for x, y in tqdm(b3_test_loader, desc="[Test Evaluation]"):
        x = x.to(device, non_blocking=True)
        with torch.cuda.amp.autocast(enabled=(device == "cuda")):
            out = b3_model(x)
        test_preds.extend(out.argmax(1).cpu().numpy())
        test_targets.extend(y.numpy())

CLASSES = ['l_pass', 'r_pass', 'l_spike', 'r_spike', 'l_set', 'r_set', 'l_winpoint', 'r_winpoint']
test_acc = accuracy_score(test_targets, test_preds)
test_f1_weighted = f1_score(test_targets, test_preds, average='weighted', zero_division=0)
test_f1_macro = f1_score(test_targets, test_preds, average='macro', zero_division=0)
test_prec = precision_score(test_targets, test_preds, average='weighted', zero_division=0)
test_rec = recall_score(test_targets, test_preds, average='weighted', zero_division=0)

metrics_dict = {
    "model": "B3 - Fine-tuned Person Classification",
    "test_accuracy": float(test_acc),
    "test_f1_weighted": float(test_f1_weighted),
    "test_f1_macro": float(test_f1_macro),
    "test_precision": float(test_prec),
    "test_recall": float(test_rec),
    "best_val_accuracy": float(best_val_acc)
}

with open(os.path.join(METRICS_DIR, "metrics.json"), 'w') as f:
    json.dump(metrics_dict, f, indent=4)

pd.DataFrame(history).to_csv(os.path.join(METRICS_DIR, "training_history.csv"), index=False)

rep_dict = classification_report(test_targets, test_preds, target_names=CLASSES, output_dict=True, zero_division=0)
pd.DataFrame(rep_dict).transpose().to_csv(os.path.join(METRICS_DIR, "classification_report.csv"))

print(f"\nTest Accuracy: {test_acc * 100:.2f}% | Test F1 (Weighted): {test_f1_weighted:.4f}\n")
print(classification_report(test_targets, test_preds, target_names=CLASSES, zero_division=0))

# Save & Display Plots
fig, axs = plt.subplots(1, 2, figsize=(14, 5))
axs[0].plot(history['train_loss'], label='Train Loss', color='royalblue')
axs[0].plot(history['val_loss'], label='Val Loss', color='crimson')
axs[0].set_title('B3 Loss Curve')
axs[0].set_xlabel('Epoch')
axs[0].set_ylabel('Loss')
axs[0].legend()
axs[0].grid(True, alpha=0.3)

axs[1].plot(history['train_acc'], label='Train Acc (%)', color='royalblue')
axs[1].plot(history['val_acc'], label='Val Acc (%)', color='crimson')
axs[1].set_title('B3 Accuracy Curve')
axs[1].set_xlabel('Epoch')
axs[1].set_ylabel('Accuracy (%)')
axs[1].legend()
axs[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig(os.path.join(PLOTS_DIR, "training_curves.png"), dpi=300)
plt.show()

cm = confusion_matrix(test_targets, test_preds)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=CLASSES, yticklabels=CLASSES)
plt.title('B3 Confusion Matrix (Test Split)')
plt.xlabel('Predicted Label')
plt.ylabel('Ground Truth Label')
plt.tight_layout()
plt.savefig(os.path.join(PLOTS_DIR, "confusion_matrix.png"), dpi=300)
plt.show()
```
