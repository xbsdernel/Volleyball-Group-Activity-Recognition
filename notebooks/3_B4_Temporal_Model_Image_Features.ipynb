```python
from tqdm.auto import tqdm
import torch
from torch import nn
import torch.optim as optim
import torchvision.transforms as transforms
from torchvision import models
from torch.utils.data import Dataset, DataLoader
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, confusion_matrix, classification_report,
)
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image
import numpy as np
import os
import random
import warnings
warnings.filterwarnings('ignore')
```

```python
# Centralized Results & Artifacts Setup
RESULTS_DIR = "results/3_B4_Temporal_Model_Image_Features"
PLOTS_DIR = os.path.join(RESULTS_DIR, "plots")
METRICS_DIR = os.path.join(RESULTS_DIR, "metrics")
os.makedirs(PLOTS_DIR, exist_ok=True)
os.makedirs(METRICS_DIR, exist_ok=True)

# â”€â”€ Reproducibility â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
SEED = 42

# â”€â”€ Training â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
LR                          = 1e-3
BATCH_SIZE                  = 2        # kept small; OOM-safe on Kaggle T4/P100
EPOCHS                      = 100
GRADIENT_ACCUMULATION_STEPS = 8       # effective batch = BATCH_SIZE Ã— GA = 16
WARMUP_EPOCHS               = 5       # linear LR warmup before cosine decay
NUM_FRAMES                  = 9       # frames per clip (middle Â± 4)

# â”€â”€ Data loading â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
NUM_WORKERS     = 2    # Kaggle provides 2-4 vCPUs
PREFETCH_FACTOR = 2

# â”€â”€ Early stopping â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
EARLY_STOP_PATIENCE  = 10     # epochs without improvement before stop
EARLY_STOP_MIN_DELTA = 0.002  # smaller gains count: the 0.5% delta stopped the LSTM head while it was still improving

# â”€â”€ Dataset class names (Volleyball dataset â€” fixed ordering) â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
CLASSES = ['r_set', 'r_spike', 'r-pass', 'r_winpoint',
           'l_set', 'l-spike', 'l-pass', 'l_winpoint']

# â”€â”€ Paths â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
DATA_ROOT        = "/kaggle/input/datasets/ahmedmohamed365/volleyball/volleyball_/videos"
RESNET_CKPT_PATH = "/kaggle/working/resnet_checkpoint.pth"
RESNET_BEST_PATH = "/kaggle/working/resnet50_best.pth"
LSTM_CKPT_PATH   = "/kaggle/working/lstm_checkpoint.pth"
LSTM_BEST_PATH   = "/kaggle/working/lstm_best.pth"
LSTM_LR                     = 3e-4   # stage-2 LR: at 1e-3 the LSTM head reached ~99% train acc in 2 epochs (overfit)
FEATURE_NOISE_STD           = 0.02   # std of Gaussian noise added to precomputed features during LSTM training (input augmentation)
```

```python
def set_seed(seed: int) -> None:
    """Seed all RNGs for reproducibility."""
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    # benchmark=True: cuDNN picks fastest kernels for fixed input size (5-15% speedup)
    # Slightly non-deterministic; set both False for exact repro if needed
    torch.backends.cudnn.benchmark     = True
    torch.backends.cudnn.deterministic = False

set_seed(SEED)

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Device : {device}")
if device == "cuda":
    print(f"GPU    : {torch.cuda.get_device_name(0)}")
    total_vram = torch.cuda.get_device_properties(0).total_memory / 1024**3
    print(f"VRAM   : {total_vram:.1f} GB")
```

```python
data = []

for video_folder_name in sorted(os.listdir(DATA_ROOT)):
    current_video_folder_path = os.path.join(DATA_ROOT, video_folder_name)

    if not os.path.isdir(current_video_folder_path):
        continue

    annotations_path = os.path.join(current_video_folder_path, "annotations.txt")
    if not os.path.isfile(annotations_path):
        continue

    vid_id = int(video_folder_name)

    with open(annotations_path, 'r') as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) >= 2:
                image_filename = parts[0]
                activity       = parts[1]
                name_folder    = image_filename.split('.')[0]
                image_path     = os.path.join(
                    current_video_folder_path, name_folder, image_filename
                )
                data.append([vid_id, image_path, activity])

print(f"Total annotated frames loaded: {len(data)}")
```

```python
def get_sequence_frame_paths(middle_frame_path: str, num_frames_around: int = 4):
    """
    Returns a list of (2*num_frames_around + 1) frame paths centred on middle_frame_path.
    Returns None if any frame is missing from disk.
    """
    clip_folder_path = os.path.dirname(middle_frame_path)
    middle_frame_num = int(os.path.splitext(os.path.basename(middle_frame_path))[0])

    sequence_paths = []
    for i in range(-num_frames_around, num_frames_around + 1):
        frame_path = os.path.join(clip_folder_path, f"{middle_frame_num + i}.jpg")
        if not os.path.isfile(frame_path):
            return None  # Incomplete sequence â€” skip this clip
        sequence_paths.append(frame_path)

    return sequence_paths

b4_clips_data = []
skipped_count = 0

for vid_id, middle_image_path, activity in data:
    seq = get_sequence_frame_paths(middle_image_path, num_frames_around=4)
    if seq is not None:
        b4_clips_data.append([vid_id, seq, activity])
    else:
        skipped_count += 1

print(f"Original annotated frames : {len(data)}")
print(f"Valid 9-frame sequences   : {len(b4_clips_data)}")
if skipped_count:
    print(f"WARNING - Skipped (missing frames): {skipped_count}")
```

```python
TRAIN_VIDEOS = {1, 3, 6, 7, 10, 13, 15, 16, 18, 22, 23, 31, 32, 36, 38, 39, 40, 41, 42, 48, 50, 52, 53, 54}
VAL_VIDEOS   = {0, 2, 8, 12, 17, 19, 24, 26, 27, 28, 30, 33, 46, 49, 51}
TEST_VIDEOS  = {4, 5, 9, 11, 14, 20, 21, 25, 29, 34, 35, 37, 43, 44, 45, 47}

train_data_b4, val_data_b4, test_data_b4 = [], [], []

for video_id, sequence_paths, activity in b4_clips_data:
    if video_id in TRAIN_VIDEOS:
        train_data_b4.append([sequence_paths, activity])
    elif video_id in VAL_VIDEOS:
        val_data_b4.append([sequence_paths, activity])
    elif video_id in TEST_VIDEOS:
        test_data_b4.append([sequence_paths, activity])

print(f"Training sequences  : {len(train_data_b4)}")
print(f"Validation sequences: {len(val_data_b4)}")
print(f"Test sequences      : {len(test_data_b4)}")
```

```python
label_encoder = LabelEncoder()
label_encoder.fit(CLASSES)  # Deterministic: always the same class-to-index mapping

num_classes = len(label_encoder.classes_)
print(f"Number of classes: {num_classes}")
print("\nClass â†’ Index mapping:")
for cls in label_encoder.classes_:
    idx = int(label_encoder.transform([cls])[0])
    print(f"  {cls:<15s}  ->  {idx}")
```

```python
# ImageNet normalisation statistics
MEAN = [0.485, 0.456, 0.406]
STD  = [0.229, 0.224, 0.225]

# Spatial transforms â€” applied with a SHARED seed across all frames in a clip
_clip_spatial = transforms.Compose([
    transforms.RandomResizedCrop(224),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(15),
])

# Colour jitter â€” applied INDEPENDENTLY per frame (intentional diversity)
_clip_color = transforms.ColorJitter(
    brightness=0.2, contrast=0.2, saturation=0.2, hue=0.1
)

# Final normalisation pipeline
_to_tensor_norm = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize(mean=MEAN, std=STD),
])

# Deterministic transform for val / test / feature extraction
val_test_transform = transforms.Compose([
    transforms.Resize(256),
    transforms.CenterCrop(224),
    transforms.ToTensor(),
    transforms.Normalize(mean=MEAN, std=STD),
])
```

```python
class CustomSequenceDataset(Dataset):
    """
    Loads fixed-length sequences of frames for group activity recognition.

    Args:
        data          : list of [sequence_paths, activity_label]
        label_encoder : fitted sklearn LabelEncoder
        is_train      : if True, applies temporally-consistent random augmentation
    """

    def __init__(self, data, label_encoder: LabelEncoder, is_train: bool = False):
        self.sequences = [item[0] for item in data]
        self.is_train  = is_train

        raw_labels  = [item[1] for item in data]
        self.labels = torch.as_tensor(
            label_encoder.transform(raw_labels), dtype=torch.long
        )

    def __len__(self) -> int:
        return len(self.sequences)

    def __getitem__(self, idx: int):
        sequence_paths = self.sequences[idx]
        frames = []

        if self.is_train:
            # One seed â†’ identical spatial params for every frame in this clip
            clip_seed = torch.randint(0, 2 ** 31, (1,)).item()
            for img_path in sequence_paths:
                img = Image.open(img_path).convert('RGB')
                torch.manual_seed(clip_seed)
                random.seed(clip_seed)
                img = _clip_spatial(img)    # consistent spatial transform
                img = _clip_color(img)      # independent colour jitter
                img = _to_tensor_norm(img)
                frames.append(img)
        else:
            for img_path in sequence_paths:
                img = Image.open(img_path).convert('RGB')
                img = val_test_transform(img)
                frames.append(img)

        return torch.stack(frames), self.labels[idx]   # (T, C, H, W), scalar
```

```python
train_dataset         = CustomSequenceDataset(train_data_b4, label_encoder, is_train=True)
val_dataset           = CustomSequenceDataset(val_data_b4,   label_encoder, is_train=False)
test_dataset          = CustomSequenceDataset(test_data_b4,  label_encoder, is_train=False)
# Deterministic training data â€” used for feature extraction only (no augmentation)
train_extract_dataset = CustomSequenceDataset(train_data_b4, label_encoder, is_train=False)

_pin = (device == "cuda")
_worker_kwargs = dict(num_workers=NUM_WORKERS, pin_memory=_pin)
if NUM_WORKERS > 0:
    _worker_kwargs['persistent_workers'] = True
    _worker_kwargs['prefetch_factor']    = PREFETCH_FACTOR

train_loader = DataLoader(
    train_dataset, batch_size=BATCH_SIZE, shuffle=True, drop_last=True, **_worker_kwargs
)
val_loader = DataLoader(
    val_dataset, batch_size=BATCH_SIZE, shuffle=False, **_worker_kwargs
)
test_loader = DataLoader(
    test_dataset, batch_size=BATCH_SIZE, shuffle=False, **_worker_kwargs
)
train_extract_loader = DataLoader(
    train_extract_dataset, batch_size=BATCH_SIZE, shuffle=False, **_worker_kwargs
)

print(f"Train batches   : {len(train_loader)}")
print(f"Val batches     : {len(val_loader)}")
print(f"Test batches    : {len(test_loader)}")
```

```python
model = models.resnet50(weights=models.ResNet50_Weights.DEFAULT)

num_features = model.fc.in_features  # 2048 for ResNet50

# Regularised head: Dropout â†’ Linear
model.fc = nn.Sequential(
    nn.Dropout(p=0.3),
    nn.Linear(num_features, num_classes),
)

# No DataParallel â€” single GPU on Kaggle; DataParallel only adds overhead
model = model.to(device)

total_params     = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"Total parameters    : {total_params:,}")
print(f"Trainable parameters: {trainable_params:,}")
if device == "cuda":
    print(f"VRAM after model load: {torch.cuda.memory_allocated() / 1024**2:.1f} MB")
```

```python
criterion = nn.CrossEntropyLoss()

# Separate parameter groups: backbone gets 10Ã— lower LR than head
backbone_params = [p for n, p in model.named_parameters() if 'fc' not in n]
head_params     = list(model.fc.parameters())

optimizer = optim.AdamW(
    [
        {'params': backbone_params, 'lr': LR * 0.1},   # 1e-4
        {'params': head_params,     'lr': LR},           # 1e-3
    ],
    weight_decay=1e-4,
)

# 5-epoch linear warmup then cosine annealing
_warmup = torch.optim.lr_scheduler.LinearLR(
    optimizer, start_factor=0.1, end_factor=1.0, total_iters=WARMUP_EPOCHS
)
_cosine = torch.optim.lr_scheduler.CosineAnnealingLR(
    optimizer, T_max=EPOCHS - WARMUP_EPOCHS, eta_min=1e-6
)
scheduler = torch.optim.lr_scheduler.SequentialLR(
    optimizer, schedulers=[_warmup, _cosine], milestones=[WARMUP_EPOCHS]
)

# AMP GradScaler â€” no-op if CUDA unavailable
scaler = torch.cuda.amp.GradScaler(enabled=(device == 'cuda'))

print(f"Backbone LR : {LR * 0.1:.1e}")
print(f"Head LR     : {LR:.1e}")
print(f"AMP enabled : {device == 'cuda'}")
```

```python
# â”€â”€ Metrics history â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
history = {
    'train_loss': [], 'val_loss': [],
    'train_acc':  [], 'val_acc':  [],
    'val_precision': [], 'val_recall': [], 'val_f1': [],
}

start_epoch        = 0
best_val_f1        = 0.0
early_stop_counter = 0

# â”€â”€ Resume from checkpoint if available â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
if os.path.exists(RESNET_CKPT_PATH):
    ckpt = torch.load(RESNET_CKPT_PATH, map_location=device)
    model.load_state_dict(ckpt['model_state_dict'])
    optimizer.load_state_dict(ckpt['optimizer_state_dict'])
    scheduler.load_state_dict(ckpt['scheduler_state_dict'])   # correct LR on resume
    start_epoch        = ckpt['epoch'] + 1
    best_val_f1        = ckpt.get('best_val_f1', 0.0)
    early_stop_counter = ckpt.get('early_stop_counter', 0)
    if 'history' in ckpt:
        history = ckpt['history']
    print(f"Resuming from epoch {start_epoch} | Best val F1: {best_val_f1:.4f}")

# â”€â”€ Main loop â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
for epoch in range(start_epoch, EPOCHS):

    # â”€â”€ Train â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
    model.train()
    total_loss_train = total_correct_train = seen_train = 0
    optimizer.zero_grad()

    train_bar = tqdm(train_loader, desc=f"Ep {epoch+1:3d}/{EPOCHS} [Train]", leave=False)
    for i, (inputs, labels) in enumerate(train_bar):
        B, T, C, H, W = inputs.shape
        inputs = inputs.view(B * T, C, H, W).to(device, non_blocking=True)
        labels = labels.to(device, non_blocking=True)

        with torch.cuda.amp.autocast(enabled=(device == 'cuda')):
            outputs = model(inputs)                          # (B*T, num_classes)
            outputs = outputs.view(B, T, -1).mean(dim=1)    # (B, num_classes) â€” temporal avg
            loss    = criterion(outputs, labels) / GRADIENT_ACCUMULATION_STEPS

        scaler.scale(loss).backward()

        if (i + 1) % GRADIENT_ACCUMULATION_STEPS == 0 or (i + 1) == len(train_loader):
            scaler.step(optimizer)
            scaler.update()
            optimizer.zero_grad()

        # Accumulate UNSCALED loss for accurate logging
        total_loss_train    += loss.item() * GRADIENT_ACCUMULATION_STEPS
        total_correct_train += (outputs.argmax(1) == labels).sum().item()
        seen_train          += labels.size(0)

        train_bar.set_postfix(
            loss=f"{total_loss_train / (i + 1):.4f}",
            acc=f"{total_correct_train / seen_train * 100:.2f}%",
        )

    avg_train_loss = total_loss_train / len(train_loader)
    avg_train_acc  = total_correct_train / seen_train * 100

    # â”€â”€ Validate â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
    model.eval()
    total_loss_val = total_correct_val = seen_val = 0
    all_preds, all_labels_list = [], []

    with torch.no_grad():
        val_bar = tqdm(val_loader, desc=f"Ep {epoch+1:3d}/{EPOCHS} [Val]  ", leave=False)
        for batch_idx, (inputs, labels) in enumerate(val_bar):
            B, T, C, H, W = inputs.shape
            inputs = inputs.view(B * T, C, H, W).to(device, non_blocking=True)
            labels = labels.to(device, non_blocking=True)

            with torch.cuda.amp.autocast(enabled=(device == 'cuda')):
                outputs  = model(inputs)
                outputs  = outputs.view(B, T, -1).mean(dim=1)
                val_loss = criterion(outputs, labels)

            preds               = outputs.argmax(1)
            total_loss_val     += val_loss.item()
            total_correct_val  += (preds == labels).sum().item()
            seen_val           += labels.size(0)
            all_preds.extend(preds.cpu().numpy())
            all_labels_list.extend(labels.cpu().numpy())

            val_bar.set_postfix(
                loss=f"{total_loss_val / (batch_idx + 1):.4f}",
                acc=f"{total_correct_val / seen_val * 100:.2f}%",
            )

    avg_val_loss = total_loss_val / len(val_loader)
    avg_val_acc  = total_correct_val / seen_val * 100
    precision    = precision_score(all_labels_list, all_preds, average='weighted', zero_division=0)
    recall       = recall_score(all_labels_list,   all_preds, average='weighted', zero_division=0)
    f1           = f1_score(all_labels_list,       all_preds, average='weighted', zero_division=0)

    scheduler.step()
    curr_lr = scheduler.get_last_lr()[0]

    # â”€â”€ Record â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
    history['train_loss'].append(round(avg_train_loss, 4))
    history['val_loss'].append(round(avg_val_loss, 4))
    history['train_acc'].append(round(avg_train_acc, 4))
    history['val_acc'].append(round(avg_val_acc, 4))
    history['val_precision'].append(round(precision, 4))
    history['val_recall'].append(round(recall, 4))
    history['val_f1'].append(round(f1, 4))

    print(
        f"Epoch {epoch+1:3d}/{EPOCHS} | "
        f"Train Loss: {avg_train_loss:.4f}  Acc: {avg_train_acc:.2f}% | "
        f"Val Loss: {avg_val_loss:.4f}  Acc: {avg_val_acc:.2f}% | "
        f"F1: {f1:.4f} | LR: {curr_lr:.2e}"
    )

    # â”€â”€ Rolling checkpoint (includes scheduler state for correct LR on resume) â”€
    torch.save({
        'epoch':                epoch,
        'model_state_dict':     model.state_dict(),
        'optimizer_state_dict': optimizer.state_dict(),
        'scheduler_state_dict': scheduler.state_dict(),
        'best_val_f1':          best_val_f1,
        'early_stop_counter':   early_stop_counter,
        'history':              history,
    }, RESNET_CKPT_PATH)

    # â”€â”€ Best-model checkpoint + early stopping â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
    if f1 > best_val_f1 + EARLY_STOP_MIN_DELTA:
        best_val_f1        = f1
        early_stop_counter = 0
        torch.save(model.state_dict(), RESNET_BEST_PATH)
        print(f"  [+] Best model saved  (val F1 = {best_val_f1:.4f})")
    else:
        early_stop_counter += 1
        print(f"  [-] No significant improvement  [{early_stop_counter}/{EARLY_STOP_PATIENCE}]")
        if early_stop_counter >= EARLY_STOP_PATIENCE:
            print(f"\n[STOP] Early stopping triggered at epoch {epoch + 1}.")
            break

    print("-" * 70)

print(f"\nStage 1 complete.  Best val F1 = {best_val_f1:.4f}")
```

```python
# â”€â”€ Free Stage 1 training objects from GPU memory â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
del optimizer, scheduler, scaler
torch.cuda.empty_cache()

if device == "cuda":
    allocated = torch.cuda.memory_allocated() / 1024**2
    reserved  = torch.cuda.memory_reserved()  / 1024**2
    print(f"VRAM allocated : {allocated:.1f} MB")
    print(f"VRAM reserved  : {reserved:.1f} MB")

# â”€â”€ Load best ResNet weights for feature extraction â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
if os.path.exists(RESNET_BEST_PATH):
    model.load_state_dict(torch.load(RESNET_BEST_PATH, map_location=device))
    print(f"\nLoaded best ResNet weights from: {RESNET_BEST_PATH}")

# â”€â”€ Replace classification head with Identity â†’ bare feature extractor â”€â”€â”€â”€â”€â”€â”€â”€
feature_extractor = model
feature_extractor.fc = nn.Identity()   # removes Dropout + Linear
feature_extractor.eval()

for param in feature_extractor.parameters():
    param.requires_grad = False

print("Feature extractor ready: frozen, eval mode, Identity head.")
```

```python
def precompute_features(dataloader, extractor, device):
    """
    Extracts frame-level features for all samples in a DataLoader.

    Returns:
        features : CPU FP16 tensor  (N, T, feature_dim)
        labels   : CPU Long tensor  (N,)
    """
    all_features, all_labels = [], []
    extractor.eval()

    with torch.no_grad():
        for inputs, labels in tqdm(dataloader, desc="Extracting features", leave=True):
            B, T, C, H, W = inputs.shape
            inputs = inputs.view(B * T, C, H, W).to(device, non_blocking=True)

            with torch.cuda.amp.autocast(enabled=(device == 'cuda')):
                feats = extractor(inputs)          # (B*T, feature_dim)

            feats = feats.view(B, T, feats.shape[1])
            all_features.append(feats.half().cpu())   # FP16 â€” halves RAM
            all_labels.append(labels.cpu())

    return torch.cat(all_features), torch.cat(all_labels)


print("Pre-computing TRAINING features (deterministic transform â€” no augmentation)...")
train_features, train_labels = precompute_features(train_extract_loader, feature_extractor, device)

print("Pre-computing VALIDATION features...")
val_features, val_labels = precompute_features(val_loader, feature_extractor, device)

print("Pre-computing TEST features...")
test_features, test_labels = precompute_features(test_loader, feature_extractor, device)

print(f"\nTrain : {train_features.shape}  dtype={train_features.dtype}")
print(f"Val   : {val_features.shape}  dtype={val_features.dtype}")
print(f"Test  : {test_features.shape}  dtype={test_features.dtype}")

total_mb = (train_features.nbytes + val_features.nbytes + test_features.nbytes) / 1024**2
print(f"Total feature RAM used: {total_mb:.1f} MB")

# Free feature extractor from GPU â€” LSTM only needs CPU-side feature tensors
del feature_extractor, model
torch.cuda.empty_cache()
if device == "cuda":
    print(f"VRAM after freeing extractor: {torch.cuda.memory_allocated() / 1024**2:.1f} MB")
```

```python
class LSTMClassifier(nn.Module):
    """
    Stacked LSTM for temporal sequence classification.

    Args:
        input_size   : feature dimension per timestep (2048 for ResNet50)
        hidden_size  : LSTM hidden state size
        num_layers   : number of stacked LSTM layers
        num_classes  : number of activity classes
        dropout_prob : dropout between LSTM layers and before the FC head
    """

    def __init__(self, input_size: int, hidden_size: int, num_layers: int,
                 num_classes: int, dropout_prob: float = 0.5):
        super().__init__()
        self.lstm    = nn.LSTM(input_size, hidden_size, num_layers,
                               batch_first=True, dropout=dropout_prob)
        self.dropout = nn.Dropout(dropout_prob)
        self.fc      = nn.Linear(hidden_size, num_classes)

    def forward(self, x):
        # x: (B, T, input_size) â€” already FP32 after casting from FP16 storage
        _, (hn, _) = self.lstm(x)     # hn: (num_layers, B, hidden_size)
        out = self.dropout(hn[-1])    # last-layer hidden state â†’ (B, hidden_size)
        return self.fc(out)           # (B, num_classes)


LSTM_HIDDEN = 256
LSTM_LAYERS = 2
input_size  = train_features.shape[2]   # 2048 for ResNet50

lstm_model = LSTMClassifier(
    input_size, LSTM_HIDDEN, LSTM_LAYERS, num_classes, dropout_prob=0.5
).to(device)

total_lstm_params = sum(p.numel() for p in lstm_model.parameters())
print(f"LSTM parameters: {total_lstm_params:,}")
print(lstm_model)
```

```python
lstm_train_ds = torch.utils.data.TensorDataset(train_features, train_labels)
lstm_val_ds   = torch.utils.data.TensorDataset(val_features,   val_labels)
lstm_test_ds  = torch.utils.data.TensorDataset(test_features,  test_labels)

# num_workers=0: in-memory data; no I/O benefit from worker processes
# pin_memory=False: no-op with num_workers=0
_lstm_kw = dict(num_workers=0, pin_memory=False)

lstm_train_loader = DataLoader(
    lstm_train_ds, batch_size=BATCH_SIZE, shuffle=True, drop_last=True, **_lstm_kw
)
lstm_val_loader   = DataLoader(lstm_val_ds,  batch_size=BATCH_SIZE, shuffle=False, **_lstm_kw)
lstm_test_loader  = DataLoader(lstm_test_ds, batch_size=BATCH_SIZE, shuffle=False, **_lstm_kw)

print(f"LSTM Train batches : {len(lstm_train_loader)}")
print(f"LSTM Val batches   : {len(lstm_val_loader)}")
print(f"LSTM Test batches  : {len(lstm_test_loader)}")
```

```python
lstm_criterion = nn.CrossEntropyLoss(label_smoothing=0.1)

lstm_optimizer = optim.AdamW(lstm_model.parameters(), lr=LSTM_LR, weight_decay=1e-2)

_lstm_warmup = torch.optim.lr_scheduler.LinearLR(
    lstm_optimizer, start_factor=0.1, end_factor=1.0, total_iters=WARMUP_EPOCHS
)
_lstm_cosine = torch.optim.lr_scheduler.CosineAnnealingLR(
    lstm_optimizer, T_max=EPOCHS - WARMUP_EPOCHS, eta_min=1e-6
)
lstm_scheduler = torch.optim.lr_scheduler.SequentialLR(
    lstm_optimizer, schedulers=[_lstm_warmup, _lstm_cosine], milestones=[WARMUP_EPOCHS]
)

# AMP GradScaler for LSTM â€” no-op if CUDA unavailable
lstm_scaler = torch.cuda.amp.GradScaler(enabled=(device == 'cuda'))

print("LSTM optimizer, scheduler, and AMP scaler ready.")
print(f"LR: {LSTM_LR:.1e} | Warmup: {WARMUP_EPOCHS} epochs | AMP: {device == 'cuda'}")
```

```python
lstm_history = {
    'train_loss': [], 'val_loss': [],
    'train_acc':  [], 'val_acc':  [],
    'val_precision': [], 'val_recall': [], 'val_f1': [],
}

lstm_start_epoch        = 0
lstm_best_val_f1        = 0.0
lstm_early_stop_counter = 0

# â”€â”€ Resume from LSTM checkpoint â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
if os.path.exists(LSTM_CKPT_PATH):
    ckpt = torch.load(LSTM_CKPT_PATH, map_location=device)
    lstm_model.load_state_dict(ckpt['model_state_dict'])
    lstm_optimizer.load_state_dict(ckpt['optimizer_state_dict'])
    lstm_scheduler.load_state_dict(ckpt['scheduler_state_dict'])
    lstm_start_epoch        = ckpt['epoch'] + 1
    lstm_best_val_f1        = ckpt.get('best_val_f1', 0.0)
    lstm_early_stop_counter = ckpt.get('early_stop_counter', 0)
    if 'history' in ckpt:
        lstm_history = ckpt['history']
    print(f"Resuming LSTM from epoch {lstm_start_epoch} | Best val F1: {lstm_best_val_f1:.4f}")

# â”€â”€ LSTM training loop â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
for epoch in range(lstm_start_epoch, EPOCHS):

    # â”€â”€ Train â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
    lstm_model.train()
    total_loss_train = total_correct_train = seen_train = 0
    lstm_optimizer.zero_grad()

    train_bar = tqdm(lstm_train_loader, desc=f"LSTM Ep {epoch+1:3d}/{EPOCHS} [Train]", leave=False)
    for i, (feats, labels) in enumerate(train_bar):
        # FP16 storage â†’ FP32 for numerical stability, then GPU
        feats  = feats.float().to(device, non_blocking=True)
        labels = labels.to(device, non_blocking=True)

        # Feature-space augmentation: Gaussian noise + random frame masking (input-level only, architecture unchanged)
        if lstm_model.training:
            feats = feats + torch.randn_like(feats) * FEATURE_NOISE_STD
            keep = (torch.rand(feats.size(0), feats.size(1), 1, device=feats.device) > 0.1).float()
            feats = feats * keep
        with torch.cuda.amp.autocast(enabled=(device == 'cuda')):
            outputs = lstm_model(feats)
            loss    = lstm_criterion(outputs, labels) / GRADIENT_ACCUMULATION_STEPS

        lstm_scaler.scale(loss).backward()

        if (i + 1) % GRADIENT_ACCUMULATION_STEPS == 0 or (i + 1) == len(lstm_train_loader):
            lstm_scaler.step(lstm_optimizer)
            lstm_scaler.update()
            lstm_optimizer.zero_grad()

        total_loss_train    += loss.item() * GRADIENT_ACCUMULATION_STEPS
        total_correct_train += (outputs.argmax(1) == labels).sum().item()
        seen_train          += labels.size(0)

        train_bar.set_postfix(
            loss=f"{total_loss_train / (i + 1):.4f}",
            acc=f"{total_correct_train / seen_train * 100:.2f}%",
        )

    avg_train_loss = total_loss_train / len(lstm_train_loader)
    avg_train_acc  = total_correct_train / seen_train * 100

    # â”€â”€ Validate â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
    lstm_model.eval()
    total_loss_val = total_correct_val = seen_val = 0
    all_preds, all_labels_list = [], []

    with torch.no_grad():
        val_bar = tqdm(lstm_val_loader, desc=f"LSTM Ep {epoch+1:3d}/{EPOCHS} [Val]  ", leave=False)
        for batch_idx, (feats, labels) in enumerate(val_bar):
            feats  = feats.float().to(device, non_blocking=True)
            labels = labels.to(device, non_blocking=True)

            with torch.cuda.amp.autocast(enabled=(device == 'cuda')):
                outputs  = lstm_model(feats)
                val_loss = lstm_criterion(outputs, labels)

            preds               = outputs.argmax(1)
            total_loss_val     += val_loss.item()
            total_correct_val  += (preds == labels).sum().item()
            seen_val           += labels.size(0)
            all_preds.extend(preds.cpu().numpy())
            all_labels_list.extend(labels.cpu().numpy())

            val_bar.set_postfix(
                loss=f"{total_loss_val / (batch_idx + 1):.4f}",
                acc=f"{total_correct_val / seen_val * 100:.2f}%",
            )

    avg_val_loss = total_loss_val / len(lstm_val_loader)
    avg_val_acc  = total_correct_val / seen_val * 100
    precision    = precision_score(all_labels_list, all_preds, average='weighted', zero_division=0)
    recall       = recall_score(all_labels_list,   all_preds, average='weighted', zero_division=0)
    f1           = f1_score(all_labels_list,       all_preds, average='weighted', zero_division=0)

    lstm_scheduler.step()
    curr_lr = lstm_scheduler.get_last_lr()[0]

    lstm_history['train_loss'].append(round(avg_train_loss, 4))
    lstm_history['val_loss'].append(round(avg_val_loss, 4))
    lstm_history['train_acc'].append(round(avg_train_acc, 4))
    lstm_history['val_acc'].append(round(avg_val_acc, 4))
    lstm_history['val_precision'].append(round(precision, 4))
    lstm_history['val_recall'].append(round(recall, 4))
    lstm_history['val_f1'].append(round(f1, 4))

    print(
        f"LSTM Ep {epoch+1:3d}/{EPOCHS} | "
        f"Train Loss: {avg_train_loss:.4f}  Acc: {avg_train_acc:.2f}% | "
        f"Val Loss: {avg_val_loss:.4f}  Acc: {avg_val_acc:.2f}% | "
        f"F1: {f1:.4f} | LR: {curr_lr:.2e}"
    )

    # â”€â”€ Rolling checkpoint â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
    torch.save({
        'epoch':                epoch,
        'model_state_dict':     lstm_model.state_dict(),
        'optimizer_state_dict': lstm_optimizer.state_dict(),
        'scheduler_state_dict': lstm_scheduler.state_dict(),
        'best_val_f1':          lstm_best_val_f1,
        'early_stop_counter':   lstm_early_stop_counter,
        'history':              lstm_history,
    }, LSTM_CKPT_PATH)

    # â”€â”€ Best-model checkpoint + early stopping â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
    if f1 > lstm_best_val_f1 + EARLY_STOP_MIN_DELTA:
        lstm_best_val_f1        = f1
        lstm_early_stop_counter = 0
        torch.save(lstm_model.state_dict(), LSTM_BEST_PATH)
        print(f"  [+] Best LSTM saved  (val F1 = {lstm_best_val_f1:.4f})")
    else:
        lstm_early_stop_counter += 1
        print(f"  [-] No significant improvement  [{lstm_early_stop_counter}/{EARLY_STOP_PATIENCE}]")
        if lstm_early_stop_counter >= EARLY_STOP_PATIENCE:
            print(f"\n[STOP] Early stopping triggered at LSTM epoch {epoch + 1}.")
            break

    print("-" * 70)

print(f"\nStage 2 complete.  Best LSTM val F1 = {lstm_best_val_f1:.4f}")
```
