# Baseline 1 (Fast & Optimized): Single Frame Image Classification

This notebook provides an optimized implementation of **Baseline 1** adopting the high-performance design from **Baseline 4**:
1. **Automatic Mixed Precision (AMP FP16)** for 2x-3x GPU speedup.
2. **cuDNN Benchmark** (`torch.backends.cudnn.benchmark = True`) for hardware kernel autotuning.
3. **High-throughput DataLoaders** with `pin_memory=True`, `persistent_workers=True`, and `prefetch_factor=2`.
4. **Differential Learning Rates** (10x smaller LR for pretrained backbone) + Warmup and Cosine Annealing.
5. **Native Single-GPU Execution** (eliminates `DataParallel` lock overhead).
6. **Bonus Lightning Mode**: Two-stage decoupled feature precomputation for sub-second epoch training.

```python
import os
import random
import warnings
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm.auto import tqdm

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader, TensorDataset
import torchvision.transforms as transforms
from torchvision import models

from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, confusion_matrix, classification_report
)

warnings.filterwarnings('ignore')
```

```python
# Reproducibility & cuDNN Autotuner
SEED = 42

def set_seed(seed: int) -> None:
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    # cuDNN autotuner picks fastest convolution kernels for fixed image size
    torch.backends.cudnn.benchmark = True
    torch.backends.cudnn.deterministic = False

set_seed(SEED)

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Device : {device}")
if device == "cuda":
    print(f"GPU    : {torch.cuda.get_device_name(0)}")
    total_vram = torch.cuda.get_device_properties(0).total_memory / 1024**3
    print(f"VRAM   : {total_vram:.1f} GB")
```

```python
# Paths & Hyperparameters
DATA_ROOT = "/kaggle/input/datasets/ahmedmohamed365/volleyball/volleyball_/videos"

# Centralized Results & Artifacts Setup
RESULTS_DIR = "results/1_B1_Image_Classification"
PLOTS_DIR = os.path.join(RESULTS_DIR, "plots")
METRICS_DIR = os.path.join(RESULTS_DIR, "metrics")
os.makedirs(PLOTS_DIR, exist_ok=True)
os.makedirs(METRICS_DIR, exist_ok=True)
MODEL_SAVE_PATH = os.path.join(RESULTS_DIR, "best_model.pth")

CLASSES = ['r_set', 'r_spike', 'r-pass', 'r_winpoint',
           'l_set', 'l-spike', 'l-pass', 'l_winpoint']

LR = 3e-4
BATCH_SIZE = 32              # Balanced batch size for single-frame ResNet-50 with AMP
EPOCHS = 60
WARMUP_EPOCHS = 5
IMAGE_SIZE = (224, 224)      # Standard ResNet-50 input resolution

NUM_WORKERS = 2
PREFETCH_FACTOR = 2
EARLY_STOP_PATIENCE = 10
EARLY_STOP_MIN_DELTA = 0.005
WEIGHT_DECAY = 1e-2         # strong L2: log shows train acc 99.9% vs val ~48% (severe overfitting)
LABEL_SMOOTHING = 0.1        # soft targets curb overconfident memorization
```

```python
# Parse Single-Frame Annotations & Split by Video IDs
data = []

for video_folder_name in sorted(os.listdir(DATA_ROOT)):
    current_video_folder_path = os.path.join(DATA_ROOT, video_folder_name)
    if not os.path.isdir(current_video_folder_path):
        continue

    annotations_path = os.path.join(current_video_folder_path, "annotations.txt")
    if not os.path.isfile(annotations_path):
        continue

    vid_id = int(video_folder_name)

    with open(annotations_path, 'r') as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) >= 2:
                image_filename = parts[0]
                activity = parts[1]
                name_folder = image_filename.split('.')[0]
                image_path = os.path.join(current_video_folder_path, name_folder, image_filename)
                data.append((vid_id, image_path, activity))

# Standard Video Splits (avoids frame leakage)
TRAIN_VIDEOS = {1, 3, 6, 7, 10, 13, 15, 16, 18, 22, 23, 31, 32, 36, 38, 39, 40, 41, 42, 48, 50, 52, 53, 54}
VAL_VIDEOS   = {0, 2, 8, 12, 17, 19, 24, 26, 27, 28, 30, 33, 46, 49, 51}
TEST_VIDEOS  = {4, 5, 9, 11, 14, 20, 21, 25, 29, 34, 35, 37, 43, 44, 45, 47}

train_data_b1 = [(p, a) for v, p, a in data if v in TRAIN_VIDEOS]
val_data_b1   = [(p, a) for v, p, a in data if v in VAL_VIDEOS]
test_data_b1  = [(p, a) for v, p, a in data if v in TEST_VIDEOS]

print(f"Total samples : {len(data)}")
print(f"Train samples : {len(train_data_b1)}")
print(f"Val samples   : {len(val_data_b1)}")
print(f"Test samples  : {len(test_data_b1)}")
```

```python
# Label Encoder & Transforms
label_encoder = LabelEncoder()
label_encoder.fit(CLASSES)
num_classes = len(label_encoder.classes_)
print("Classes:", label_encoder.classes_)

train_transform = transforms.Compose([
    transforms.RandomResizedCrop(IMAGE_SIZE, scale=(0.7, 1.0)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.ColorJitter(brightness=0.3, contrast=0.3, saturation=0.3, hue=0.05),
    transforms.RandomRotation(degrees=10),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
    transforms.RandomErasing(p=0.25, scale=(0.02, 0.15)),
])

val_test_transform = transforms.Compose([
    transforms.Resize(IMAGE_SIZE),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])
```

```python
# Dataset & Optimized DataLoaders
class CustomImageDataset(Dataset):
    def __init__(self, data, label_encoder, transform=None):
        self.image_paths = [item[0] for item in data]
        self.transform = transform
        raw_labels = [item[1] for item in data]
        self.labels = torch.as_tensor(label_encoder.transform(raw_labels), dtype=torch.long)

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):
        image = Image.open(self.image_paths[idx]).convert('RGB')
        if self.transform:
            image = self.transform(image)
        return image, self.labels[idx]

train_dataset = CustomImageDataset(train_data_b1, label_encoder, transform=train_transform)
val_dataset   = CustomImageDataset(val_data_b1,   label_encoder, transform=val_test_transform)
test_dataset  = CustomImageDataset(test_data_b1,  label_encoder, transform=val_test_transform)
# Deterministic train set for feature extraction
train_extract_dataset = CustomImageDataset(train_data_b1, label_encoder, transform=val_test_transform)

_loader_kwargs = dict(
    num_workers=NUM_WORKERS,
    pin_memory=(device == "cuda"),
    persistent_workers=(NUM_WORKERS > 0),
    prefetch_factor=PREFETCH_FACTOR if NUM_WORKERS > 0 else None
)

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, drop_last=True, **_loader_kwargs)
val_loader   = DataLoader(val_dataset,   batch_size=BATCH_SIZE, shuffle=False, **_loader_kwargs)
test_loader  = DataLoader(test_dataset,  batch_size=BATCH_SIZE, shuffle=False, **_loader_kwargs)
train_extract_loader = DataLoader(train_extract_dataset, batch_size=BATCH_SIZE, shuffle=False, **_loader_kwargs)

print(f"Train batches: {len(train_loader)} | Val batches: {len(val_loader)} | Test batches: {len(test_loader)}")
```

```python
# Model Definition: Pretrained ResNet-50 with Dropout Head
model = models.resnet50(weights=models.ResNet50_Weights.DEFAULT)
num_features = model.fc.in_features

# Regularized classification head
model.fc = nn.Sequential(
    nn.Dropout(p=0.5),
    nn.Linear(num_features, num_classes)
)

# Direct single-GPU placement (avoids DataParallel thread lock overhead)
model = model.to(device)

total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"Total parameters    : {total_params:,}")
print(f"Trainable parameters: {trainable_params:,}")
```

```python
# Optimizer, Differential Learning Rates, LR Scheduler & AMP Scaler
criterion = nn.CrossEntropyLoss(label_smoothing=LABEL_SMOOTHING)

# Differential learning rates: 10x lower for pretrained backbone
backbone_params = [p for n, p in model.named_parameters() if 'fc' not in n]
head_params = list(model.fc.parameters())

optimizer = optim.AdamW([
    {'params': backbone_params, 'lr': LR * 0.1},  # 3e-5
    {'params': head_params,     'lr': LR}          # 3e-4
], weight_decay=WEIGHT_DECAY)

# Warmup + Cosine Annealing scheduler
_warmup = torch.optim.lr_scheduler.LinearLR(optimizer, start_factor=0.1, end_factor=1.0, total_iters=WARMUP_EPOCHS)
_cosine = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=EPOCHS - WARMUP_EPOCHS, eta_min=1e-6)
scheduler = torch.optim.lr_scheduler.SequentialLR(optimizer, schedulers=[_warmup, _cosine], milestones=[WARMUP_EPOCHS])

# AMP GradScaler (enables FP16 Tensor Core execution)
scaler = torch.cuda.amp.GradScaler(enabled=(device == "cuda"))

print(f"AMP enabled : {device == 'cuda'}")
print(f"Backbone LR : {LR * 0.1:.1e} | Head LR: {LR:.1e}")
```

```python
# High-Speed Training Loop with AMP and Early Stopping
history = {'train_loss': [], 'val_loss': [], 'train_acc': [], 'val_acc': [], 'val_f1': []}
best_val_f1 = 0.0
early_stop_counter = 0

for epoch in range(EPOCHS):
    # --- Train Phase ---
    model.train()
    total_loss_train = 0.0
    correct_train = 0
    total_train = 0

    train_bar = tqdm(train_loader, desc=f"Epoch {epoch+1:02d}/{EPOCHS} [Train]", leave=False)
    for inputs, labels in train_bar:
        inputs = inputs.to(device, non_blocking=True)
        labels = labels.to(device, non_blocking=True)

        optimizer.zero_grad()

        # Mixed Precision Forward Pass
        with torch.cuda.amp.autocast(enabled=(device == "cuda")):
            outputs = model(inputs)
            loss = criterion(outputs, labels)

        # Scaled Backward Pass
        scaler.scale(loss).backward()
        scaler.step(optimizer)
        scaler.update()

        total_loss_train += loss.item() * inputs.size(0)
        preds = outputs.argmax(dim=1)
        correct_train += (preds == labels).sum().item()
        total_train += labels.size(0)

        train_bar.set_postfix(
            loss=f"{total_loss_train / total_train:.4f}",
            acc=f"{correct_train / total_train * 100:.2f}%"
        )

    avg_train_loss = total_loss_train / total_train
    avg_train_acc = correct_train / total_train * 100

    # --- Validation Phase ---
    model.eval()
    total_loss_val = 0.0
    correct_val = 0
    total_val = 0
    all_val_preds, all_val_labels = [], []

    with torch.no_grad():
        for inputs, labels in val_loader:
            inputs = inputs.to(device, non_blocking=True)
            labels = labels.to(device, non_blocking=True)

            with torch.cuda.amp.autocast(enabled=(device == "cuda")):
                outputs = model(inputs)
                loss = criterion(outputs, labels)

            total_loss_val += loss.item() * inputs.size(0)
            preds = outputs.argmax(dim=1)
            correct_val += (preds == labels).sum().item()
            total_val += labels.size(0)

            all_val_preds.extend(preds.cpu().numpy())
            all_val_labels.extend(labels.cpu().numpy())

    avg_val_loss = total_loss_val / total_val
    avg_val_acc = correct_val / total_val * 100
    val_f1 = f1_score(all_val_labels, all_val_preds, average='weighted', zero_division=0)

    scheduler.step()

    history['train_loss'].append(avg_train_loss)
    history['val_loss'].append(avg_val_loss)
    history['train_acc'].append(avg_train_acc)
    history['val_acc'].append(avg_val_acc)
    history['val_f1'].append(val_f1)

    print(f"Epoch {epoch+1:02d}/{EPOCHS:02d} | Train Loss: {avg_train_loss:.4f} Acc: {avg_train_acc:.2f}% | Val Loss: {avg_val_loss:.4f} Acc: {avg_val_acc:.2f}% F1: {val_f1:.4f}")

    # Checkpoint best weights
    if val_f1 > (best_val_f1 + EARLY_STOP_MIN_DELTA):
        best_val_f1 = val_f1
        early_stop_counter = 0
        torch.save(model.state_dict(), MODEL_SAVE_PATH)
        print(f"  --> Saved new best checkpoint (Val F1: {best_val_f1:.4f})")
    else:
        early_stop_counter += 1
        if early_stop_counter >= EARLY_STOP_PATIENCE:
            print(f"\n[Early Stopping] No improvement for {EARLY_STOP_PATIENCE} consecutive epochs.")
            break
```

```python
# Test Split Evaluation & Metrics Export
if os.path.exists(MODEL_SAVE_PATH):
    model.load_state_dict(torch.load(MODEL_SAVE_PATH, map_location=device))
    print(f"Loaded best weights from {MODEL_SAVE_PATH}")

model.eval()
test_preds, test_targets = [], []

with torch.no_grad():
    for inputs, labels in test_loader:
        inputs = inputs.to(device, non_blocking=True)
        with torch.cuda.amp.autocast(enabled=(device == "cuda")):
            outputs = model(inputs)
        preds = outputs.argmax(dim=1)
        test_preds.extend(preds.cpu().numpy())
        test_targets.extend(labels.numpy())

import pandas as pd, json
test_acc = accuracy_score(test_targets, test_preds)
test_f1_weighted = f1_score(test_targets, test_preds, average='weighted', zero_division=0)
test_f1_macro = f1_score(test_targets, test_preds, average='macro', zero_division=0)
test_prec = precision_score(test_targets, test_preds, average='weighted', zero_division=0)
test_rec = recall_score(test_targets, test_preds, average='weighted', zero_division=0)

metrics_dict = {
    "model": "B1 - Image Classification",
    "test_accuracy": float(test_acc),
    "test_f1_weighted": float(test_f1_weighted),
    "test_f1_macro": float(test_f1_macro),
    "test_precision": float(test_prec),
    "test_recall": float(test_rec),
    "best_val_f1": float(best_val_f1)
}

with open(os.path.join(METRICS_DIR, "metrics.json"), 'w') as f:
    json.dump(metrics_dict, f, indent=4)

pd.DataFrame(history).to_csv(os.path.join(METRICS_DIR, "training_history.csv"), index=False)

rep_dict = classification_report(test_targets, test_preds, target_names=label_encoder.classes_, output_dict=True, zero_division=0)
pd.DataFrame(rep_dict).transpose().to_csv(os.path.join(METRICS_DIR, "classification_report.csv"))

print(f"\nTest Accuracy: {test_acc * 100:.2f}% | Test F1 (Weighted): {test_f1_weighted:.4f}\n")
print(classification_report(test_targets, test_preds, target_names=label_encoder.classes_))
```

```python
# Plot & Save Training Metrics and Confusion Matrix
fig, axs = plt.subplots(1, 2, figsize=(14, 5))

axs[0].plot(history['train_loss'], label='Train Loss', color='royalblue')
axs[0].plot(history['val_loss'], label='Val Loss', color='crimson')
axs[0].set_title('Loss Curve')
axs[0].set_xlabel('Epoch')
axs[0].set_ylabel('Loss')
axs[0].legend()
axs[0].grid(True, alpha=0.3)

axs[1].plot(history['train_acc'], label='Train Acc', color='royalblue')
axs[1].plot(history['val_acc'], label='Val Acc', color='crimson')
axs[1].plot([f * 100 for f in history['val_f1']], label='Val F1 (%)', color='forestgreen', linestyle='--')
axs[1].set_title('Accuracy & F1 Curve')
axs[1].set_xlabel('Epoch')
axs[1].set_ylabel('Percentage (%)')
axs[1].legend()
axs[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig(os.path.join(PLOTS_DIR, "training_curves.png"), dpi=300)
plt.show()

cm = confusion_matrix(test_targets, test_preds)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=label_encoder.classes_, yticklabels=label_encoder.classes_)
plt.title('B1 Confusion Matrix (Test Split)')
plt.xlabel('Predicted Label')
plt.ylabel('Ground Truth Label')
plt.tight_layout()
plt.savefig(os.path.join(PLOTS_DIR, "confusion_matrix.png"), dpi=300)
plt.show()
```

## Bonus: Lightning Mode (Decoupled Feature Precomputation — The B4 Secret)

Extract all 2048-dim feature vectors **once** using the backbone, store them in RAM as half-precision FP16, delete the CNN from GPU memory, and train the classifier in **sub-second epochs**.

```python
# 1. Feature Extractor Setup
feature_extractor = model
feature_extractor.fc = nn.Identity()
feature_extractor.eval()
for param in feature_extractor.parameters():
    param.requires_grad = False

def extract_features(loader, desc="Extracting"):
    feats, labels = [], []
    with torch.no_grad():
        for imgs, lbls in tqdm(loader, desc=desc):
            imgs = imgs.to(device, non_blocking=True)
            with torch.cuda.amp.autocast(enabled=(device == "cuda")):
                f = feature_extractor(imgs)
            feats.append(f.half().cpu())   # FP16 in CPU RAM
            labels.append(lbls.cpu())
    return torch.cat(feats), torch.cat(labels)

print("Precomputing features offline once...")
train_feats, train_lbls = extract_features(train_extract_loader, "Train Features")
val_feats,   val_lbls   = extract_features(val_loader,           "Val Features")
test_feats,  test_lbls  = extract_features(test_loader,          "Test Features")

# Delete CNN from GPU memory completely
del feature_extractor, model
torch.cuda.empty_cache()
print("CNN backbone removed from GPU VRAM.")

# 2. Fast In-Memory Classifier Training
fast_train_loader = DataLoader(TensorDataset(train_feats, train_lbls), batch_size=128, shuffle=True)
fast_val_loader   = DataLoader(TensorDataset(val_feats,   val_lbls),   batch_size=128, shuffle=False)

fast_classifier = nn.Sequential(
    nn.Dropout(p=0.3),
    nn.Linear(2048, num_classes)
).to(device)

fast_optimizer = optim.AdamW(fast_classifier.parameters(), lr=3e-4, weight_decay=1e-2)
fast_criterion = nn.CrossEntropyLoss()

print("\n--- Training Fast Classifier on Precomputed Features ---")
for ep in range(30):
    fast_classifier.train()
    for x, y in fast_train_loader:
        x, y = x.float().to(device), y.to(device)
        fast_optimizer.zero_grad()
        loss = fast_criterion(fast_classifier(x), y)
        loss.backward()
        fast_optimizer.step()

    if (ep + 1) % 5 == 0:
        fast_classifier.eval()
        corr, tot = 0, 0
        with torch.no_grad():
            for x, y in fast_val_loader:
                x, y = x.float().to(device), y.to(device)
                corr += (fast_classifier(x).argmax(1) == y).sum().item()
                tot += y.size(0)
        print(f"Fast Head Epoch {ep+1:02d}/30 | Val Acc: {corr / tot * 100:.2f}%")
```
