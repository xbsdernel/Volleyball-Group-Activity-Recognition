# Model 6: Baseline 7 (B7) â€” Two-stage Model without LSTM 2
**Reference:** Ibrahim et al., *A Hierarchical Deep Temporal Model for Group Activity Recognition*, CVPR 2016 (Section 4.1 #7)

### Architecture & Workflow
This is a variant of the full hierarchical model that **removes LSTM 2** (the group-level temporal model). Instead, the final hidden states of individual person LSTMs (LSTM 1) are max-pooled directly to classify group activity, without any secondary group temporal modeling.

**Data Flow:**
$$\underbrace{12 \text{ tracklets} \times 9 \text{ frames}}_{\text{input}} \xrightarrow{\text{Person CNN}} (B{\times}K, 9, 2048) \xrightarrow{\text{LSTM 1}} (B{\times}K, 9, h_1) \xrightarrow{h_T} (B, K, h_1) \xrightarrow{\max_k} (B, h_1) \xrightarrow{\text{Linear}} 8 \text{ classes}$$

**Key difference from full Hierarchical Model:** No LSTM 2 â€” the max-pooled LSTM 1 outputs feed directly into the group activity classifier.
**Key difference from B6:** B6 has LSTM 2 but no LSTM 1. B7 has LSTM 1 but no LSTM 2.

```python
import os
import json
import random
import warnings
from PIL import Image
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm.auto import tqdm

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import torchvision.transforms as transforms
from torchvision import models
from sklearn.metrics import accuracy_score, f1_score, precision_score, recall_score, confusion_matrix, classification_report

warnings.filterwarnings('ignore')

VIDEOS_ROOT   = "/kaggle/input/datasets/ahmedmohamed365/volleyball/volleyball_/videos"
TRACKING_ROOT = "/kaggle/input/datasets/ahmedmohamed365/volleyball/volleyball_tracking_annotation/volleyball_tracking_annotation"

# Results directory setup
RESULTS_DIR = "results/6_B7_Two_Stage_No_LSTM2"
PLOTS_DIR = os.path.join(RESULTS_DIR, "plots")
METRICS_DIR = os.path.join(RESULTS_DIR, "metrics")
os.makedirs(PLOTS_DIR, exist_ok=True)
os.makedirs(METRICS_DIR, exist_ok=True)

SEED = 42
random.seed(SEED); np.random.seed(SEED)
torch.manual_seed(SEED); torch.cuda.manual_seed_all(SEED)
torch.backends.cudnn.benchmark = True

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Device: {device}")
```

```python
TRAIN_VIDEOS = {1, 3, 6, 7, 10, 13, 15, 16, 18, 22, 23, 31, 32, 36, 38, 39, 40, 41, 42, 48, 50, 52, 53, 54}
VAL_VIDEOS   = {0, 2, 8, 12, 17, 19, 24, 26, 27, 28, 30, 33, 46, 49, 51}
TEST_VIDEOS  = {4, 5, 9, 11, 14, 20, 21, 25, 29, 34, 35, 37, 43, 44, 45, 47}

LABEL_MAP = {
    "l_pass": 0, "r_pass": 1, "l_spike": 2, "r_spike": 3,
    "l_set": 4, "r_set": 5, "l_winpoint": 6, "r_winpoint": 7,
    "l-pass": 0, "r-pass": 1, "l-spike": 2, "r-spike": 3,
    "l-set": 4, "r-set": 5, "l-winpoint": 6, "r-winpoint": 7
}
PERSON_ACTION_MAP = {
    "waiting": 0, "setting": 1, "digging": 2, "falling": 3,
    "spiking": 4, "blocking": 5, "jumping": 6, "moving": 7, "standing": 8
}

IMAGE_SIZE = (224, 224); BATCH_SIZE = 8; LR = 1e-3
EPOCHS_STAGE1 = 15; EPOCHS_STAGE2 = 40
NUM_FRAMES = 9; NUM_PLAYERS = 12; LSTM1_HIDDEN = 1024
```

```python
# Tracking & Utility Functions
class BoxInfo:
    def __init__(self, line):
        words = line.split(); self.category = words.pop()
        words = [int(w) for w in words]; self.player_ID = words[0]
        x1, y1, x2, y2, frame_ID, *_ = words[1:]
        self.box = (x1, y1, x2, y2); self.frame_ID = frame_ID

def load_tracking_annot(path):
    player_boxes = {i: [] for i in range(12)}; frame_boxes_dct = {}
    if not os.path.isfile(path): return frame_boxes_dct
    with open(path, 'r') as f:
        for line in f:
            b = BoxInfo(line)
            if b.player_ID > 11: continue
            player_boxes[b.player_ID].append(b)
    for pid, boxes in player_boxes.items():
        boxes = boxes[5:-6] if len(boxes) > 11 else boxes
        for b in boxes: frame_boxes_dct.setdefault(b.frame_ID, []).append(b)
    return frame_boxes_dct

def get_seq_paths(mid_path, n=4):
    d = os.path.dirname(mid_path); name, ext = os.path.splitext(os.path.basename(mid_path))
    mid = int(name)
    return [p if os.path.isfile(p := os.path.join(d, f"{mid+o}{ext}")) else mid_path for o in range(-n, n+1)]
```

```python
# Load Full Dataset Annotations
data = []
for vid_name in sorted(os.listdir(VIDEOS_ROOT)):
    vpath = os.path.join(VIDEOS_ROOT, vid_name)
    if not os.path.isdir(vpath): continue
    ann_path = os.path.join(vpath, "annotations.txt")
    if not os.path.isfile(ann_path): continue
    vid_id = int(vid_name)
    with open(ann_path) as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) >= 2:
                img_name, lbl = parts[0], parts[1]
                clip_id = os.path.splitext(img_name)[0]
                mid_path = os.path.join(vpath, clip_id, img_name)
                trk_path = os.path.join(TRACKING_ROOT, str(vid_id), clip_id, f"{clip_id}.txt")
                data.append({"video_id": vid_id, "frame_paths": get_seq_paths(mid_path),
                              "tracking_path": trk_path, "group_label": lbl, "image_path": mid_path})

train_data = [d for d in data if d["video_id"] in TRAIN_VIDEOS]
val_data   = [d for d in data if d["video_id"] in VAL_VIDEOS]
test_data  = [d for d in data if d["video_id"] in TEST_VIDEOS]
print(f"Total clips: {len(data)} | Train: {len(train_data)} | Val: {len(val_data)} | Test: {len(test_data)}")
```

```python
crop_transform = transforms.Compose([
    transforms.Resize(IMAGE_SIZE), transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

# Stage 1: Person action crops dataset
class PersonCropDataset(Dataset):
    def __init__(self, data, transform=None):
        self.samples = []; self.transform = transform
        for d in data:
            if not os.path.isfile(d.get("image_path", "")) or not os.path.isfile(d["tracking_path"]): continue
            fboxes = load_tracking_annot(d["tracking_path"])
            if not fboxes: continue
            fid = list(fboxes.keys())[0]
            for b in fboxes[fid][:12]:
                self.samples.append((d["image_path"], b.box, PERSON_ACTION_MAP.get(b.category.lower(), 8)))

    def __len__(self): return len(self.samples)
    def __getitem__(self, idx):
        p, (x1,y1,x2,y2), act = self.samples[idx]
        img = Image.open(p).convert("RGB"); w, h = img.size
        c = img.crop((max(0,x1), max(0,y1), min(w,max(x1+1,x2)), min(h,max(y1+1,y2))))
        if self.transform: c = self.transform(c)
        return c, torch.tensor(act, dtype=torch.long)

# Stage 2: 12-player tracklet dataset (12, 9, 3, 224, 224) per sample
class HierarchicalDataset(Dataset):
    def __init__(self, data, transform=None):
        self.data = data; self.transform = transform

    def __len__(self): return len(self.data)
    def __getitem__(self, idx):
        d = self.data[idx]
        fboxes = load_tracking_annot(d["tracking_path"])
        # Build per-player tracklets: all_players[k] = list of 9 crops for player k
        all_players = [[] for _ in range(NUM_PLAYERS)]
        for p in d["frame_paths"]:
            img = Image.open(p).convert("RGB"); w, h = img.size
            fid_str = os.path.splitext(os.path.basename(p))[0]
            try: fid = int(fid_str)
            except: fid = list(fboxes.keys())[0] if fboxes else 0
            boxes = fboxes.get(fid, [])
            for k in range(NUM_PLAYERS):
                if k < len(boxes):
                    x1,y1,x2,y2 = boxes[k].box
                    crop = img.crop((max(0,x1), max(0,y1), min(w,max(x1+1,x2)), min(h,max(y1+1,y2))))
                else:
                    crop = Image.new("RGB", IMAGE_SIZE, 0)
                if self.transform: crop = self.transform(crop)
                all_players[k].append(crop)
        # Stack: (12, 9, 3, 224, 224)
        player_tracklets = torch.stack([torch.stack(trk) for trk in all_players])
        label = torch.tensor(LABEL_MAP[d["group_label"]], dtype=torch.long)
        return player_tracklets, label
```

```python
# Stage 1: Fine-tune Person CNN + LSTM 1 on person action labels
class PersonLSTM1(nn.Module):
    def __init__(self, num_actions=9, hidden_dim=1024, dropout=0.5):
        super().__init__()
        resnet = models.resnet50(weights=models.ResNet50_Weights.DEFAULT)
        self.backbone = nn.Sequential(*list(resnet.children())[:-1])
        self.lstm1 = nn.LSTM(input_size=2048, hidden_size=hidden_dim, num_layers=1, batch_first=True)
        self.dropout = nn.Dropout(dropout)
        self.action_head = nn.Linear(hidden_dim, num_actions)

    def forward_features(self, tracklet):
        """tracklet: (B, 9, 3, 224, 224) -> lstm1_out: (B, 9, hidden_dim), last_h: (B, hidden_dim)"""
        B, T, C, H, W = tracklet.shape
        feats = self.backbone(tracklet.view(B*T, C, H, W)).view(B, T, 2048)
        lstm_out, (hn, _) = self.lstm1(feats)
        return lstm_out, hn[-1]  # (B, 9, hidden_dim), (B, hidden_dim)

    def forward(self, tracklet):
        lstm_out, _ = self.forward_features(tracklet)
        return self.action_head(self.dropout(lstm_out))  # (B, 9, num_actions)

person_lstm1 = PersonLSTM1(num_actions=9, hidden_dim=LSTM1_HIDDEN).to(device)

# Stage 1 would be trained on individual player tracklets with action labels
# Skipping Stage 1 training loop here (see B3 notebook for person action fine-tuning pattern)
# torch.save(person_lstm1.state_dict(), "/kaggle/working/person_lstm1_b7.pth")
print(f"PersonLSTM1 parameters: {sum(p.numel() for p in person_lstm1.parameters()):,}")
```

```python
# Stage 2: B7 Model â€” LSTM 1 Representations Max-pooled -> Group Classifier (NO LSTM 2)
class B7Model(nn.Module):
    """
    B7: Two-stage model WITHOUT LSTM 2.
    Person CNN + LSTM 1 -> max-pool last hidden state over players -> Group Linear classifier.
    LSTM 1 frozen after Stage 1 training.
    """
    def __init__(self, person_lstm1, num_classes=8, hidden_dim=1024, dropout=0.5):
        super().__init__()
        self.person_lstm1 = person_lstm1
        for p in self.person_lstm1.parameters():
            p.requires_grad = False  # Frozen Stage 1
        self.dropout = nn.Dropout(dropout)
        self.classifier = nn.Linear(hidden_dim, num_classes)

    def forward(self, x):
        # x: (B, 12, 9, 3, 224, 224)
        B, K, T, C, H, W = x.shape
        x_flat = x.view(B * K, T, C, H, W)
        with torch.no_grad():
            _, last_h = self.person_lstm1.forward_features(x_flat)  # (B*K, hidden_dim)
        # Reshape to (B, K, hidden_dim) and max-pool over players
        last_h = last_h.view(B, K, -1)   # (B, 12, hidden_dim)
        group_repr = last_h.max(dim=1)[0]  # (B, hidden_dim) â€” max over 12 players
        return self.classifier(self.dropout(group_repr))

b7_model = B7Model(person_lstm1=person_lstm1, num_classes=8, hidden_dim=LSTM1_HIDDEN).to(device)
trainable_params = sum(p.numel() for p in b7_model.parameters() if p.requires_grad)
print(f"B7 trainable parameters (classifier only): {trainable_params:,}")
```

```python
# DataLoaders
b7_train_ds = HierarchicalDataset(train_data, transform=crop_transform)
b7_val_ds   = HierarchicalDataset(val_data,   transform=crop_transform)
b7_test_ds  = HierarchicalDataset(test_data,  transform=crop_transform)

loader_kw = dict(num_workers=2, pin_memory=(device == "cuda"), persistent_workers=True)
b7_train_loader = DataLoader(b7_train_ds, batch_size=BATCH_SIZE, shuffle=True, drop_last=True, **loader_kw)
b7_val_loader   = DataLoader(b7_val_ds,   batch_size=BATCH_SIZE, shuffle=False, **loader_kw)
b7_test_loader  = DataLoader(b7_test_ds,  batch_size=BATCH_SIZE, shuffle=False, **loader_kw)
print(f"Train batches: {len(b7_train_loader)} | Val batches: {len(b7_val_loader)} | Test batches: {len(b7_test_loader)}")
```
