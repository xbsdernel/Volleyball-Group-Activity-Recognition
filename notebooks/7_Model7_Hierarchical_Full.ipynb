# Model 7: Two-stage Hierarchical Model â€” Full Architecture
**Reference:** Ibrahim et al., *A Hierarchical Deep Temporal Model for Group Activity Recognition*, CVPR 2016 (Section 3, Figure 2, Equations 7 & 8)

## Architecture Overview
This is the **complete proposed model** from the paper. It is a two-stage, end-to-end hierarchical deep temporal architecture.

**Stage 1 â€” Person-Level Temporal Modeling:**
- A ResNet-50 CNN (replacing AlexNet per agreed backbone substitution) extracts visual features $x_{tk}$ for each player $k$ at each timestep $t$.
- A person-level LSTM (LSTM 1, 1024 hidden units) models each individual player's action dynamics over 9 frames, producing $h_{tk}$.
- LSTM 1 is trained on person-level action labels (9 categories) in Stage 1.

**Stage 2 â€” Group-Level Temporal Modeling:**
- For each player $k$ at each frame $t$, the CNN feature $x_{tk}$ and LSTM 1 hidden state $h_{tk}$ are concatenated:
$$P_{tk} = \text{concat}(x_{tk},\, h_{tk}) \quad [\text{Eq. 7}]$$
- Max-pooled over all 12 players per frame:
$$Z_t = \max_{k=1..12} P_{tk} \quad [\text{Eq. 8}]$$
- A fully-connected projection layer (3000 units) then feeds the sequence of 9 $Z_t$ vectors into a group LSTM (LSTM 2, 500 hidden units).
- A final linear softmax layer classifies into 8 group activity classes.

> **Permitted deviation:** ResNet-50 is used in place of AlexNet as explicitly agreed. All other components, dimensionalities, and data flows follow the paper exactly.

```python
import os
import json
import random
import warnings
from PIL import Image
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm.auto import tqdm

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import torchvision.transforms as transforms
from torchvision import models
from sklearn.metrics import accuracy_score, f1_score, precision_score, recall_score, confusion_matrix, classification_report

warnings.filterwarnings('ignore')

VIDEOS_ROOT   = "/kaggle/input/datasets/ahmedmohamed365/volleyball/volleyball_/videos"
TRACKING_ROOT = "/kaggle/input/datasets/ahmedmohamed365/volleyball/volleyball_tracking_annotation/volleyball_tracking_annotation"

# Results directory setup
RESULTS_DIR = "results/7_Model7_Hierarchical_Full"
PLOTS_DIR = os.path.join(RESULTS_DIR, "plots")
METRICS_DIR = os.path.join(RESULTS_DIR, "metrics")
os.makedirs(PLOTS_DIR, exist_ok=True)
os.makedirs(METRICS_DIR, exist_ok=True)

SEED = 42
random.seed(SEED); np.random.seed(SEED)
torch.manual_seed(SEED); torch.cuda.manual_seed_all(SEED)
torch.backends.cudnn.benchmark = True

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Device: {device}")
if device == "cuda":
    print(f"GPU: {torch.cuda.get_device_name(0)}")
    print(f"VRAM: {torch.cuda.get_device_properties(0).total_memory / 1024**3:.1f} GB")
```

```python
# Official Dataset Configuration
TRAIN_VIDEOS = {1, 3, 6, 7, 10, 13, 15, 16, 18, 22, 23, 31, 32, 36, 38, 39, 40, 41, 42, 48, 50, 52, 53, 54}
VAL_VIDEOS   = {0, 2, 8, 12, 17, 19, 24, 26, 27, 28, 30, 33, 46, 49, 51}
TEST_VIDEOS  = {4, 5, 9, 11, 14, 20, 21, 25, 29, 34, 35, 37, 43, 44, 45, 47}

LABEL_MAP = {
    "l_pass": 0, "r_pass": 1, "l_spike": 2, "r_spike": 3,
    "l_set": 4, "r_set": 5, "l_winpoint": 6, "r_winpoint": 7,
    "l-pass": 0, "r-pass": 1, "l-spike": 2, "r-spike": 3,
    "l-set": 4, "r-set": 5, "l-winpoint": 6, "r-winpoint": 7
}
PERSON_ACTION_MAP = {
    "waiting": 0, "setting": 1, "digging": 2, "falling": 3,
    "spiking": 4, "blocking": 5, "jumping": 6, "moving": 7, "standing": 8
}

IMAGE_SIZE    = (224, 224)
NUM_FRAMES    = 9
NUM_PLAYERS   = 12
NUM_CLASSES   = 8
NUM_ACTIONS   = 9

# Hyperparameters from paper / common practice
LSTM1_HIDDEN  = 1024  # Person LSTM hidden dim (paper does not specify exactly; 1024 is used)
PROJ_DIM      = 3000  # FC projection layer after max-pooling (Paper Section 3.3)
LSTM2_HIDDEN  = 500   # Group LSTM hidden dim (Paper Section 3.3)

LR            = 1e-3
BATCH_SIZE    = 4     # Large tensors: (4, 12, 9, 3, 224, 224)
EPOCHS_S1     = 15
EPOCHS_S2     = 40
```

```python
# Tracking Parser & Utilities
class BoxInfo:
    def __init__(self, line):
        words = line.split(); self.category = words.pop()
        words = [int(w) for w in words]; self.player_ID = words[0]
        x1, y1, x2, y2, frame_ID, *_ = words[1:]
        self.box = (x1, y1, x2, y2); self.frame_ID = frame_ID

def load_tracking_annot(path):
    player_boxes = {i: [] for i in range(12)}; frame_boxes_dct = {}
    if not os.path.isfile(path): return frame_boxes_dct
    with open(path) as f:
        for line in f:
            b = BoxInfo(line)
            if b.player_ID > 11: continue
            player_boxes[b.player_ID].append(b)
    for pid, boxes in player_boxes.items():
        boxes = boxes[5:-6] if len(boxes) > 11 else boxes
        for b in boxes: frame_boxes_dct.setdefault(b.frame_ID, []).append(b)
    return frame_boxes_dct

def get_seq_paths(mid_path, n=4):
    d = os.path.dirname(mid_path); name, ext = os.path.splitext(os.path.basename(mid_path))
    mid = int(name)
    return [p if os.path.isfile(p := os.path.join(d, f"{mid+o}{ext}")) else mid_path for o in range(-n, n+1)]
```

```python
# Load Full Dataset (All 55 Videos)
data = []
for vid_name in sorted(os.listdir(VIDEOS_ROOT)):
    vpath = os.path.join(VIDEOS_ROOT, vid_name)
    if not os.path.isdir(vpath): continue
    ann_path = os.path.join(vpath, "annotations.txt")
    if not os.path.isfile(ann_path): continue
    vid_id = int(vid_name)
    with open(ann_path) as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) >= 2:
                img_name, lbl = parts[0], parts[1]
                clip_id = os.path.splitext(img_name)[0]
                mid_path = os.path.join(vpath, clip_id, img_name)
                trk_path = os.path.join(TRACKING_ROOT, str(vid_id), clip_id, f"{clip_id}.txt")
                data.append({"video_id": vid_id, "frame_paths": get_seq_paths(mid_path),
                              "tracking_path": trk_path, "group_label": lbl, "image_path": mid_path})

train_data = [d for d in data if d["video_id"] in TRAIN_VIDEOS]
val_data   = [d for d in data if d["video_id"] in VAL_VIDEOS]
test_data  = [d for d in data if d["video_id"] in TEST_VIDEOS]
print(f"Total clips: {len(data)} | Train: {len(train_data)} | Val: {len(val_data)} | Test: {len(test_data)}")
```

```python
crop_transform = transforms.Compose([
    transforms.Resize(IMAGE_SIZE), transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

# Stage 1: Per-player tracklet dataset
class PersonTrackletDataset(Dataset):
    """Returns (9, 3, 224, 224) tracklet + action label for Stage 1 person LSTM training."""
    def __init__(self, data, transform=None):
        self.samples = []; self.transform = transform
        for d in data:
            if not os.path.isfile(d.get("image_path", "")) or not os.path.isfile(d["tracking_path"]): continue
            fboxes = load_tracking_annot(d["tracking_path"])
            if not fboxes: continue
            fid = list(fboxes.keys())[0]
            for b in fboxes[fid][:NUM_PLAYERS]:
                act_idx = PERSON_ACTION_MAP.get(b.category.lower(), 8)
                self.samples.append((d["frame_paths"], b.player_ID, act_idx, d["tracking_path"]))

    def __len__(self): return len(self.samples)
    def __getitem__(self, idx):
        frame_paths, pid, act_idx, trk_path = self.samples[idx]
        fboxes = load_tracking_annot(trk_path)
        crops = []
        for p in frame_paths:
            img = Image.open(p).convert("RGB"); w, h = img.size
            fid_str = os.path.splitext(os.path.basename(p))[0]
            try: fid = int(fid_str)
            except: fid = list(fboxes.keys())[0] if fboxes else 0
            boxes = fboxes.get(fid, [])
            box = next((b for b in boxes if b.player_ID == pid), None)
            if box:
                x1,y1,x2,y2 = box.box
                c = img.crop((max(0,x1), max(0,y1), min(w,max(x1+1,x2)), min(h,max(y1+1,y2))))
            else:
                c = Image.new("RGB", IMAGE_SIZE, 0)
            if self.transform: c = self.transform(c)
            crops.append(c)
        return torch.stack(crops), torch.tensor(act_idx, dtype=torch.long)  # (9,3,H,W), scalar

# Stage 2: Full hierarchical dataset
class HierarchicalDataset(Dataset):
    """Returns (12, 9, 3, 224, 224) player tracklets + group label for Stage 2."""
    def __init__(self, data, transform=None):
        self.data = data; self.transform = transform

    def __len__(self): return len(self.data)
    def __getitem__(self, idx):
        d = self.data[idx]
        fboxes = load_tracking_annot(d["tracking_path"])
        all_players = [[] for _ in range(NUM_PLAYERS)]
        for p in d["frame_paths"]:
            img = Image.open(p).convert("RGB"); w, h = img.size
            fid_str = os.path.splitext(os.path.basename(p))[0]
            try: fid = int(fid_str)
            except: fid = list(fboxes.keys())[0] if fboxes else 0
            boxes = fboxes.get(fid, [])
            for k in range(NUM_PLAYERS):
                if k < len(boxes):
                    x1,y1,x2,y2 = boxes[k].box
                    crop = img.crop((max(0,x1), max(0,y1), min(w,max(x1+1,x2)), min(h,max(y1+1,y2))))
                else:
                    crop = Image.new("RGB", IMAGE_SIZE, 0)
                if self.transform: crop = self.transform(crop)
                all_players[k].append(crop)
        # (12, 9, 3, 224, 224)
        player_tracklets = torch.stack([torch.stack(trk) for trk in all_players])
        label = torch.tensor(LABEL_MAP[d["group_label"]], dtype=torch.long)
        return player_tracklets, label
```

```python
# ============================================================
# Full Two-Stage Hierarchical Model (Paper: Figure 2, Eq. 7-8)
# ============================================================
class TwoStageHierarchicalModel(nn.Module):
    """
    Full Two-Stage Hierarchical Deep Temporal Model.
    Ibrahim et al., CVPR 2016 â€” Section 3, Figure 2.

    Stage 1 (Person-level):
      Person CNN (ResNet-50 backbone, replacing AlexNet per agreed substitution)
      + LSTM 1 (1024 hidden units) trained on person action labels.

    Stage 2 (Group-level):
      Concat(x_tk, h_tk) per player per frame  [Eq. 7]
      -> Max-pool over K=12 players: Z_t        [Eq. 8]
      -> FC projection (3000 units)
      -> LSTM 2 (500 hidden units)
      -> Linear (8 group activity classes)
    """
    def __init__(self, h1=1024, proj=3000, h2=500, num_classes=8, num_actions=9, dropout=0.5):
        super().__init__()

        # === Stage 1 Components ===
        resnet = models.resnet50(weights=models.ResNet50_Weights.DEFAULT)
        self.person_cnn = nn.Sequential(*list(resnet.children())[:-1])  # -> (B, 2048, 1, 1)

        self.lstm1 = nn.LSTM(input_size=2048, hidden_size=h1, num_layers=1, batch_first=True)
        self.person_action_head = nn.Linear(h1, num_actions)  # Trained in Stage 1

        # === Stage 2 Components ===
        concat_dim = 2048 + h1  # x_tk concat h_tk
        self.fc_proj = nn.Sequential(
            nn.Linear(concat_dim, proj),
            nn.ReLU(inplace=True),
            nn.Dropout(dropout)
        )
        self.lstm2 = nn.LSTM(input_size=proj, hidden_size=h2, num_layers=1, batch_first=True)
        self.dropout = nn.Dropout(dropout)
        self.group_classifier = nn.Linear(h2, num_classes)

    # --- Stage 1 Forward ---
    def forward_stage1(self, tracklet):
        """tracklet: (B, 9, 3, 224, 224) -> action logits (B, 9, num_actions)"""
        B, T, C, H, W = tracklet.shape
        feats = self.person_cnn(tracklet.view(B*T, C, H, W)).view(B, T, 2048)  # (B, 9, 2048)
        lstm1_out, _ = self.lstm1(feats)  # (B, 9, h1)
        return self.person_action_head(lstm1_out)  # (B, 9, num_actions)

    # --- Stage 2 Forward ---
    def forward_stage2(self, x):
        """x: (B, 12, 9, 3, 224, 224) -> group logits (B, num_classes)"""
        B, K, T, C, H, W = x.shape
        x_flat = x.view(B * K, T, C, H, W)  # Process all player tracklets together

        # 1. CNN features x_tk
        with torch.no_grad():
            cnn_feats = self.person_cnn(x_flat.view(B*K*T, C, H, W)).view(B*K, T, 2048)

        # 2. LSTM 1 hidden states h_tk
        lstm1_out, _ = self.lstm1(cnn_feats)  # (B*K, 9, h1)

        # 3. Concatenation: P_tk = concat(x_tk, h_tk)  [Eq. 7]
        p_tk = torch.cat([cnn_feats, lstm1_out], dim=-1)  # (B*K, 9, 2048+h1)
        p_tk = p_tk.view(B, K, T, -1).permute(0, 2, 1, 3)  # (B, 9, 12, concat_dim)

        # 4. Max-pool over K=12 players: Z_t = max_k(P_tk)  [Eq. 8]
        z_t = p_tk.max(dim=2)[0]  # (B, 9, concat_dim)

        # 5. FC projection (3000 units)
        z_proj = self.fc_proj(z_t)  # (B, 9, proj_dim)

        # 6. Group LSTM 2 (500 units)
        _, (hn, _) = self.lstm2(z_proj)
        h_group = hn[-1]  # (B, h2)

        # 7. Group classification
        return self.group_classifier(self.dropout(h_group))  # (B, num_classes)

    def forward(self, x):
        return self.forward_stage2(x)

hier_model = TwoStageHierarchicalModel(
    h1=LSTM1_HIDDEN, proj=PROJ_DIM, h2=LSTM2_HIDDEN,
    num_classes=NUM_CLASSES, num_actions=NUM_ACTIONS
).to(device)

total_params = sum(p.numel() for p in hier_model.parameters())
print(f"Total parameters: {total_params:,}")
print(f"LSTM1 hidden: {LSTM1_HIDDEN} | FC proj: {PROJ_DIM} | LSTM2 hidden: {LSTM2_HIDDEN}")
```

### Stage 1: Train Person CNN + LSTM 1 on Person Action Labels

```python
stage1_train_ds = PersonTrackletDataset(train_data, transform=crop_transform)
stage1_val_ds   = PersonTrackletDataset(val_data, transform=crop_transform)
stage1_loader   = DataLoader(stage1_train_ds, batch_size=16, shuffle=True, num_workers=2, pin_memory=True, persistent_workers=True)
print(f"Stage 1 tracklet samples -> Train: {len(stage1_train_ds)} | Val: {len(stage1_val_ds)}")

# Stage 1 trains only the person CNN + LSTM 1 + action head
stage1_params = list(hier_model.person_cnn.parameters()) + \
                list(hier_model.lstm1.parameters()) + \
                list(hier_model.person_action_head.parameters())
optimizer1 = optim.AdamW(stage1_params, lr=1e-4, weight_decay=1e-4)
criterion1 = nn.CrossEntropyLoss()
scaler1    = torch.cuda.amp.GradScaler(enabled=(device == "cuda"))

print(f"Stage 1 trainable params: {sum(p.numel() for p in stage1_params):,}")
print("Uncomment training loop below to train Stage 1:")
# for epoch in range(EPOCHS_S1):
#     hier_model.train()
#     correct, total = 0, 0
#     for tracklets, action_labels in tqdm(stage1_loader, desc=f"Stage1 Ep{epoch+1}", leave=False):
#         tracklets = tracklets.to(device, non_blocking=True)
#         action_labels = action_labels.to(device, non_blocking=True)
#         optimizer1.zero_grad()
#         with torch.cuda.amp.autocast(enabled=(device == "cuda")):
#             logits = hier_model.forward_stage1(tracklets)  # (B, 9, 9)
#             # Flatten time dimension: treat each timestep as a separate prediction
#             B, T, C_act = logits.shape
#             loss = criterion1(logits.view(B*T, C_act), action_labels.unsqueeze(1).expand(B, T).reshape(B*T))
#         scaler1.scale(loss).backward()
#         scaler1.step(optimizer1); scaler1.update()
#         correct += (logits[:, -1, :].argmax(1) == action_labels).sum().item(); total += B
#     print(f"Stage1 Ep{epoch+1} | Action Acc: {correct/total*100:.2f}%")
# torch.save(hier_model.state_dict(), "/kaggle/working/hier_stage1.pth")
```

### Stage 2: Train Group-Level Temporal Modeling (FC Proj + LSTM 2 + Classifier)
Person CNN and LSTM 1 are frozen. Only Stage 2 components are trained on group activity labels.

```python
# Load Stage 1 weights if available
stage1_ckpt = "/kaggle/working/hier_stage1.pth"
if os.path.isfile(stage1_ckpt):
    hier_model.load_state_dict(torch.load(stage1_ckpt, map_location=device))
    print(f"Loaded Stage 1 weights from {stage1_ckpt}")

# Freeze Stage 1 components (Person CNN + LSTM 1) â€” Stage 2 only trains FC proj + LSTM 2 + classifier
for p in hier_model.person_cnn.parameters():      p.requires_grad = False
for p in hier_model.lstm1.parameters():           p.requires_grad = False
for p in hier_model.person_action_head.parameters(): p.requires_grad = False

stage2_params = list(hier_model.fc_proj.parameters()) + \
                list(hier_model.lstm2.parameters()) + \
                list(hier_model.group_classifier.parameters())
print(f"Stage 2 trainable params: {sum(p.numel() for p in stage2_params):,}")

# DataLoaders
s2_train_ds = HierarchicalDataset(train_data, transform=crop_transform)
s2_val_ds   = HierarchicalDataset(val_data,   transform=crop_transform)
s2_test_ds  = HierarchicalDataset(test_data,  transform=crop_transform)

loader_kw = dict(num_workers=2, pin_memory=(device == "cuda"), persistent_workers=True)
s2_train_loader = DataLoader(s2_train_ds, batch_size=BATCH_SIZE, shuffle=True, drop_last=True, **loader_kw)
s2_val_loader   = DataLoader(s2_val_ds,   batch_size=BATCH_SIZE, shuffle=False, **loader_kw)
s2_test_loader  = DataLoader(s2_test_ds,  batch_size=BATCH_SIZE, shuffle=False, **loader_kw)
print(f"Stage 2 -> Train batches: {len(s2_train_loader)} | Val: {len(s2_val_loader)} | Test: {len(s2_test_loader)}")
```

```python
# Stage 2 Training Loop
optimizer2 = optim.AdamW(stage2_params, lr=LR, weight_decay=1e-4)
criterion2 = nn.CrossEntropyLoss()
scaler2    = torch.cuda.amp.GradScaler(enabled=(device == "cuda"))
scheduler2 = optim.lr_scheduler.CosineAnnealingLR(optimizer2, T_max=EPOCHS_S2, eta_min=1e-6)

history = {"train_loss": [], "val_loss": [], "train_acc": [], "val_acc": []}
best_val_acc = 0.0

for epoch in range(EPOCHS_S2):
    # --- Train ---
    hier_model.train()
    total_loss, correct, total = 0.0, 0, 0
    for inputs, labels in tqdm(s2_train_loader, desc=f"Ep {epoch+1:02d}/{EPOCHS_S2} [Train Hier]", leave=False):
        inputs, labels = inputs.to(device, non_blocking=True), labels.to(device, non_blocking=True)
        optimizer2.zero_grad()
        with torch.cuda.amp.autocast(enabled=(device == "cuda")):
            out = hier_model.forward_stage2(inputs)
            loss = criterion2(out, labels)
        scaler2.scale(loss).backward()
        scaler2.step(optimizer2); scaler2.update()
        total_loss += loss.item() * inputs.size(0)
        correct += (out.argmax(1) == labels).sum().item(); total += labels.size(0)
    tr_loss, tr_acc = total_loss / total, correct / total * 100

    # --- Validate ---
    hier_model.eval(); total_loss, correct, total = 0.0, 0, 0
    with torch.no_grad():
        for inputs, labels in tqdm(s2_val_loader, desc=f"Ep {epoch+1:02d} [Val]", leave=False):
            inputs, labels = inputs.to(device, non_blocking=True), labels.to(device, non_blocking=True)
            with torch.cuda.amp.autocast(enabled=(device == "cuda")):
                out = hier_model.forward_stage2(inputs)
                loss = criterion2(out, labels)
            total_loss += loss.item() * inputs.size(0)
            correct += (out.argmax(1) == labels).sum().item(); total += labels.size(0)
    val_loss, val_acc = total_loss / total, correct / total * 100
    scheduler2.step()

    history["train_loss"].append(tr_loss); history["val_loss"].append(val_loss)
    history["train_acc"].append(tr_acc);   history["val_acc"].append(val_acc)
    print(f"Ep {epoch+1:02d}/{EPOCHS_S2} | Train Loss: {tr_loss:.4f} Acc: {tr_acc:.2f}% | Val Loss: {val_loss:.4f} Acc: {val_acc:.2f}%")

    if val_acc > best_val_acc:
        best_val_acc = val_acc
        torch.save(hier_model.state_dict(), "/kaggle/working/hierarchical_best.pth")
torch.save(hier_model.state_dict(), os.path.join(RESULTS_DIR, "best_model.pth"))
        print(f"  --> Saved best hierarchical model checkpoint (Val Acc: {best_val_acc:.2f}%)")
```

```python
# Final Test Evaluation & Metrics Export
best_ckpt = os.path.join(RESULTS_DIR, "best_model.pth")
if not os.path.isfile(best_ckpt):
    best_ckpt = "/kaggle/working/hierarchical_best.pth"
if os.path.isfile(best_ckpt):
    hier_model.load_state_dict(torch.load(best_ckpt, map_location=device))

hier_model.eval()
preds, targets = [], []
with torch.no_grad():
    for inputs, labels in tqdm(s2_test_loader, desc="[Test]"):
        inputs = inputs.to(device, non_blocking=True)
        with torch.cuda.amp.autocast(enabled=(device == "cuda")):
            out = hier_model.forward_stage2(inputs)
        preds.extend(out.argmax(1).cpu().numpy()); targets.extend(labels.numpy())

class_names = ["l_pass", "r_pass", "l_spike", "r_spike", "l_set", "r_set", "l_winpoint", "r_winpoint"]
test_acc = accuracy_score(targets, preds)
test_f1_weighted = f1_score(targets, preds, average='weighted', zero_division=0)
test_f1_macro = f1_score(targets, preds, average='macro', zero_division=0)
test_prec = precision_score(targets, preds, average='weighted', zero_division=0)
test_rec = recall_score(targets, preds, average='weighted', zero_division=0)

metrics_dict = {
    "model": "Model #7 - Two-stage Hierarchical Model (Full)",
    "test_accuracy": float(test_acc),
    "test_f1_weighted": float(test_f1_weighted),
    "test_f1_macro": float(test_f1_macro),
    "test_precision": float(test_prec),
    "test_recall": float(test_rec),
    "best_val_accuracy": float(best_val_acc)
}

with open(os.path.join(METRICS_DIR, "metrics.json"), 'w') as f:
    json.dump(metrics_dict, f, indent=4)

pd.DataFrame(history).to_csv(os.path.join(METRICS_DIR, "training_history.csv"), index=False)

rep_dict = classification_report(targets, preds, target_names=class_names, output_dict=True, zero_division=0)
pd.DataFrame(rep_dict).transpose().to_csv(os.path.join(METRICS_DIR, "classification_report.csv"))

print(f"\nTest Accuracy: {test_acc*100:.2f}%")
print(f"Test F1 (weighted): {test_f1_weighted:.4f}\n")
print(classification_report(targets, preds, target_names=class_names, zero_division=0))
```
